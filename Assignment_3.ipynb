# Factors and Arbitrage

We assume that stock returns follow a factor structure with two factors $f_1$ and $f_2$. Furthermore, the risk-free rate $r_f = 3\%$, and $A$ and $B$ are two well-diversified portfolios with the following properties:

\begin{center}
\begin{tabular}{lrrr}
\toprule
Portfolio & $\beta_1$ & $\beta_2$ & $E[r]$ \\
\midrule
$A$ & 1 & 0.5 & 7.5\% \\
$B$ & 0.2 & 2 & 13.4\% \\
\bottomrule
\end{tabular}
\end{center}

---

## (a) The price of risk for each factor, $f_1$ and $f_2$, under no arbitrage:

For a well-diversified portfolio idiosyncratic risk is diversified away, so expected excess returns are fully explained by the factor exposure,

$$E[r_i] - r_f = \lambda_1 \beta_{i1} + \lambda_2 \beta_{i2},$$

where $\lambda_n$ is the excess return per unit of exposure to factor $n$.

Substituting from the initial table yields the following two equations:

$$
\begin{aligned}
A: && 0.075 - 0.03 &= \lambda_1 \cdot 1 + \lambda_2 \cdot 0.5\\
B: && 0.134 - 0.03 &= \lambda_1 \cdot 0.2 + \lambda_2 \cdot 2
\end{aligned}
$$

We find the price of risk for each factor by solving equation $A$ for $\lambda_1$ and substituting into $B$:

$$
\begin{aligned}
A: && \lambda_1 &= 0.045 - 0.5\lambda_2\\
B: && 0.104 &= (0.045 - 0.5\lambda_2)\cdot 0.2 + \lambda_2 \cdot 2\\
&& 0.104 &= 0.009 + 1.9\lambda_2\\
&& \lambda_2 &= 0.05 \quad \Longrightarrow \quad \lambda_1 = 0.02
\end{aligned}
$$

The price of risk for factor $f_1$ is $\lambda_1 = 2.0\%$ and for factor $f_2$ it is $\lambda_2 = 5.0\%$, in this system.

---

## (b) Identifying and exploiting an arbitrage opportunity:

$C$ is a well-diversified portfolio with factor loadings $(\beta_{C1}, \beta_{C2}) = (0.5,\ 1.0)$ and expected return $E^{\text{obs}}[r_C] = 9.5\%$:

\begin{center}
\begin{tabular}{lrrr}
\toprule
Portfolio & $\beta_{C1}$ & $\beta_{C2}$ & $E^{\text{obs}}[r]$ \\
\midrule
$C$ & 0.5 & 1.0 & 9.5\% \\
\bottomrule
\end{tabular}
\end{center}

According to $C$'s factor loadings and the prices of risk calculated in (a), the expected no-arbitrage return of $C$ is

$$
\begin{aligned}
E^{\text{model}}[r_C] &= 0.03 + \lambda_1 \cdot 0.5 + \lambda_2 \cdot 1\\
&= 9.0\%
\end{aligned}
$$

Because the value does not coincide with the observed return in the table, there exists an arbitrage opportunity.

To take advantage of it, we create a portfolio $H$ made up of $A$, $B$ and the risk-free asset, which replicates the factor exposure of $C$:

$$
\begin{aligned}
&& 1 &= w_A + w_B + w_f\\
f_1: && 0.5 &= w_A + 0.2\,w_B + 0\,w_f\\
f_2: && 1 &= 0.5\,w_A + 2\,w_B + 0\,w_f
\end{aligned}
$$

Solving equation $f_1$ for the weight in portfolio $B$ and substituting into equation $f_2$ yields the respective weights in portfolio $H$:

$$
\begin{aligned}
f_1: && w_B &= 2.5 - 5w_A\\
f_2: && 1 &= 0.5 w_A + 2 \cdot (2.5 - 5 w_A)\\
&& 1 &= 5 - 9.5w_A \\
&& 4 &= 9.5\,w_A\\
&& w_A &= \tfrac{16}{38} \quad \Longrightarrow \quad w_B = \tfrac{15}{38} \quad \Longrightarrow \quad w_f = \tfrac{7}{38}
\end{aligned}
$$

The return of this newly created portfolio $H$ is

$$
\begin{aligned}
E[r_H] &= w_A\,E[r_A] + w_B\,E[r_B] + w_f\, r_f \\
E[r_H] &= \tfrac{16}{38}(7.5\%) + \tfrac{15}{38}(13.4\%) + \tfrac{7}{38}(3\%) = 9.0\%
\end{aligned}
$$

As a consistency check, $E[r_H] = 9.0\% = r_f + 0.5\lambda_1 + 1.0\lambda_2 = E^{\text{model}}[r_C]$, which confirms that $H$ replicates $C$’s factor exposures.

Therefore, we would take a long position in portfolio $C$ and an equally sized short position in $H$. The net investment is zero, the net exposure to each factor is zero, and as such the position earns a riskless profit of $9.5\% - 9.0\% = 0.5\%$ per unit of capital.

# Performance measures

Consider the following properties for two portfolios:

\begin{center}
\begin{tabular}{lrr}
\toprule
Portfolio & $E[r^e]$ & $\sigma$ \\
\midrule
Market ($M$) & 10\% & 25\% \\
$P$ & 15\% & 35\% \\
\bottomrule
\end{tabular}
\end{center}

The Sharpe ratios of the two portfolios are

$$S_M = \frac{E[r^e_M]}{\sigma_M} = \frac{0.10}{0.25} = 0.4000, \qquad S_P = \frac{E[r^e_P]}{\sigma_P} = \frac{0.15}{0.35} \approx 0.4286 .$$

---

## (a) Constructing $P^*$

In the $(\sigma, E[r])$ plane, all weight allocations between a risky portfolio and the risk-free asset lie on a line through $(0, r_f)$, with the slope representing that portfolio's Sharpe ratio. By moving along this linear function, we can adjust $(\sigma, E[r])$, while the Sharpe ratio of $P$ remains constant. In this example, $\sigma_P$ needs to be lowered to match $\sigma_M$. We therefore move left, since $\sigma_M < \sigma_P$.

Let the investor hold a weight $w_P$ in portfolio $P$ and $w_f = 1 - w_P$ in the risk-free asset. The excess return of the combined portfolio is

$$
\begin{aligned}
E[r_{P^*}] &= w_P\,E[r_P] + (1-w_P)r_f \\
E[r^e_{P^*}] &= w_P\, E[r^e_P]
\end{aligned}
$$

since the risk-free asset contributes no excess return. The portfolio volatility is

$$
\begin{aligned}
\sigma_{P^*} &= \sqrt{w_P^2 \sigma_P^2 + w_f^2 \sigma_f^2 + 2 w_P w_f \sigma_{P,f}} \\
\sigma_{P^*} &= \sqrt{w_P^2 \sigma_P^2} = w_P\, \sigma_P,
\end{aligned}
$$

since the risk-free asset has zero variance and zero covariance with $P$.

We see that both $E[r^e_{P^*}]$ and $\sigma_{P^*}$ solely relate to the original levels of $P$ scaled by the same factor $w_P$. It follows that the Sharpe ratio is unaffected by the allocation between $P$ and the risk-free asset, and formalizes the initial "linear function" understanding.

By imposing the matching volatility constraint we can derive the required weight in portfolio $P$:

$$
\begin{aligned}
\sigma_{P^*} &= w_P\, \sigma_P = \sigma_M \\
\Longrightarrow \qquad w_P &= \frac{\sigma_M}{\sigma_P} = \frac{0.25}{0.35} = \frac{5}{7} \approx 0.7143 \\
\Longrightarrow \qquad w_f &= 1 - w_P = \frac{2}{7} \approx 0.2857
\end{aligned}
$$

Substituting into the excess-return expression above yields the expected excess return for $P^{*}$:

$$E[r^e_{P^*}] = w_P\, E[r^e_P] = \tfrac{5}{7} \cdot 15\% \approx 10.71\%$$

The portfolio $P^*$ that matches the volatility of the market therefore invests $71.4\%$ in $P$ and $28.6\%$ in the risk-free asset.

---

## (b) Equivalence of the two comparisons

Following the results from (a), combining $P$ with the risk-free asset scales the mean and volatility by the same factor, and $P^*$ inherits $P$’s Sharpe ratio:

$$S_{P^*} = \frac{E[r^e_{P^*}]}{\sigma_{P^*}} = \frac{w_P\, E[r^e_P]}{w_P\, \sigma_P} = \frac{E[r^e_P]}{\sigma_P} = S_P$$

The Sharpe ratio is independent of the allocation between $P$ and the risk-free asset as the scaling term cancels out.

Using this fact, $S_P = S_{P^*}$, and the initial constraint $\sigma_{P^*} = \sigma_M$, both terms can be rewritten to have a common denominator:

$$S_P - S_M = S_{P^*} - S_M = \frac{E[r^e_{P^*}]}{\sigma_{P^*}} - \frac{E[r^e_M]}{\sigma_M} = \frac{E[r^e_{P^*}] - E[r^e_M]}{\sigma_M}$$

The two comparisons therefore rank $P$ and $M$ identically. This can also be numerically verified as the comparison yields

$$S_P - S_M = 0.4286 - 0.4000 = 0.0286, \qquad \frac{E[r^e_{P^*}] - E[r^e_M]}{\sigma_M} = \frac{10.71\% - 10.00\%}{25\%} = 0.0286 .$$

Comparing $E[r^e_{P^*}]$ with $E[r^e_M]$ is the more intuitive of the two. It answers how much additional expected return the investor earns by choosing $P$ over the market portfolio, while holding volatility fixed at the market's level of $25\%$. In this example, the investor earns $0.71$ percentage points more per year by choosing $P$ over $M$, which is economically concrete.

The Sharpe ratio comparison has the opposite advantage, that it is applied to the two initial portfolios, without the need to construct a synthetic portfolio $P^*$, and it is unaffected by leverage.

---

## (c) Optimal combination of $P$ and $M$

Portfolio $P$ offers a higher Sharpe ratio than the market, and a market beta of 1.2. By using a single-factor model, we can separate $P$'s excess return into alpha, market exposure, and residual risk with no expected excess return:

$$r^e_P = \alpha_P + \beta_P\, r^e_M + \varepsilon_P, \qquad E[\varepsilon_P] = 0, \qquad \mathrm{Cov}(\varepsilon_P, r^e_M) = 0$$

The alpha of $P$ is the expected excess return not explained by its market exposure:

$$\alpha_P = E[r^e_P] - \beta_P\, E[r^e_M] = 0.15 - 1.2 \cdot 0.10 = 0.03 = 3\%$$

The total portfolio variance consists of a systematic component, scaled by the squared market beta, and an idiosyncratic component. Solving for the idiosyncratic part yields:

$$\sigma_\varepsilon^2 = \sigma_P^2 - \beta_P^2 \sigma_M^2 = 0.35^2 - 1.2^2 \cdot 0.25^2 = 0.1225 - 0.0900 = 0.0325,$$

$$\sigma_\varepsilon = \sqrt{0.0325} \approx 0.1803 = 18.03\%, \qquad \mathrm{Cov}(r_P, r_M) = \beta_P \sigma_M^2 = 1.2 \cdot 0.0625 = 0.075 .$$

To find the optimal weight $w_P^0$ in the unadjusted active position, we compare its alpha per unit of residual variance with the market's reward per unit of market variance:

$$w_P^0 = \frac{\alpha_P / \sigma_\varepsilon^2}{E[r^e_M] / \sigma_M^2} = \frac{0.03 / 0.0325}{0.10 / 0.0625} = \frac{0.9231}{1.6000} = \frac{15}{26} \approx 0.5769 .$$

Because $P$ is not beta-neutral, a position in $P$ also carries market exposure. We therefore adjust the weight by the portfolio's market beta

$$w_P = \frac{w_P^0}{1 + (1 - \beta_P)\, w_P^0} = \frac{15/26}{1 - 0.2 \cdot 15/26} = \frac{15/26}{23/26} = \frac{15}{23} \approx 0.6522$$

$$w_M = 1 - w_P = \frac{8}{23} \approx 0.3478 .$$

The optimal risky portfolio $P_{\text{opt}}$ therefore holds $65.2\%$ in $P$ and $34.8\%$ in the market portfolio. Since $\beta_P > 1$, the denominator of the adjustment is smaller than one and $w_P > w_P^0$. The resulting market exposure of the combined portfolio is

$$\beta_{\text{opt}} = w_P \beta_P + w_M = \tfrac{15}{23} \cdot 1.2 + \tfrac{8}{23} \approx 1.13 .$$

---

## (d) Sharpe ratio of the optimal portfolio

The squared Sharpe ratio of the optimal portfolio consists of the squared Sharpe ratio of the market and the squared contribution of the active position $\alpha_P / \sigma_\varepsilon$, measured as alpha per unit of residual risk:

$$S_{\text{opt}}^2 = S_M^2 + \left( \frac{\alpha_P}{\sigma_\varepsilon} \right)^2$$

Substituting the respective values yields a Sharpe ratio

$$\frac{\alpha_P}{\sigma_\varepsilon} = \frac{0.03}{0.1803} \approx 0.1664$$

$$S_{\text{opt}}^2 = 0.4000^2 + 0.1664^2 = 0.1600 + 0.0277 = 0.1877 \qquad \Longrightarrow \qquad S_{\text{opt}} = \sqrt{0.1877} \approx 0.4332 .$$

Alternatively, we can compute the excess return of $P_{\text{opt}}$ and its variance to find its Sharpe ratio:

$$E[r^e_{\text{opt}}] = \tfrac{15}{23} \cdot 0.15 + \tfrac{8}{23} \cdot 0.10 = 0.1326$$

and

$$\sigma_{\text{opt}}^2 = w_P^2 \sigma_P^2 + w_M^2 \sigma_M^2 + 2 w_P w_M \, \mathrm{Cov}(r_P, r_M)$$

$$\sigma_{\text{opt}}^2 = \left( \tfrac{15}{23} \right)^2 (0.1225) + \left( \tfrac{8}{23} \right)^2 (0.0625) + 2 \left( \tfrac{15}{23} \right) \left( \tfrac{8}{23} \right) (0.075) \approx 0.0937$$

$$\sigma_{\text{opt}} = \sqrt{0.0937} \approx 0.3061 .$$

Calculating the Sharpe ratio now, arrives at the same level as before:

$$S_{\text{opt}} = \frac{E[r^e_{\text{opt}}]}{\sigma_{\text{opt}}} = \frac{0.1326}{0.3061} \approx 0.4332.$$

The optimal portfolio has a Sharpe ratio of $0.4332$, which exceeds both $S_P \approx 0.4286$ and $S_M = 0.4000$, confirming that combining the two portfolios dominates holding either one in isolation.

## Data and monthly excess returns

For each fund we use the monthly *adjusted* closing price from Yahoo Finance. The adjusted price assumes that all dividends and capital gains distributions are reinvested, so the simple monthly return
$$
r_{i,t} = \frac{P^{adj}_{i,t}}{P^{adj}_{i,t-1}} - 1
$$
is the fund's total return. Because a fund's net asset value is reported after the deduction of the management fee and other expenses, these are net-of-fee returns, i.e. the returns that an investor in the fund actually earned. The price series starts in December 2001 so that the first return is the one for January 2002. Yahoo labels each monthly bar with the first day of the month (e.g. 2026-07-01), while the price itself is the closing price on the last trading day of that month, so the bar dated 2026-07-01 delivers the July 2026 return.

From the Kenneth French data library we use the monthly file *Fama/French 3 Factors* and the *Momentum Factor*, for the factors and risk-free rate:

- $r_{f,t}$ (RF): one-month Treasury bill return,
- $r_{M,t}-r_{f,t}$ (Mkt-RF): excess return of the value-weighted market portfolio of all CRSP firms listed on the NYSE, AMEX and NASDAQ,
- $SMB_t$ (Small Minus Big): return on small-cap stocks minus large-cap stocks (size factor),
- $HML_t$ (High Minus Low): return on high book-to-market (value) stocks minus low book-to-market (growth) stocks (value factor),
- $Mom_t$: return on stocks with high prior 2-12 month returns minus stocks with low prior returns (momentum factor of Carhart, 1997).

The file reports returns in percent, so all values are divided by 100. SMB, HML and Mom are returns on zero-investment long-short portfolios and are therefore already "excess" returns.

After matching both data sets by calendar month, the monthly excess return of fund $i$ is
$$
r^e_{i,t} = r_{i,t} - r_{f,t}.
$$

The code below downloads the fund prices from Yahoo Finance and the factors from the Kenneth French data library and saves them as `fund_prices_yahoo.csv` and `ffc4_factors_french.csv` next to the notebook. If the two files already exist, they are loaded instead of downloaded again. The results below are based on data downloaded on 13 October 2026.

```python
DATA_DIR = Path.cwd()
PRICE_FILE = DATA_DIR / "fund_prices_yahoo.csv"
FACTOR_FILE = DATA_DIR / "ffc4_factors_french.csv"
FRENCH_URL = "https://mba.tuck.dartmouth.edu/pages/faculty/ken.french/ftp/"

PRICE_START, PRICE_END = "2001-12-01", "2026-08-01"            # Yahoo end date is exclusive
SAMPLE_START, SAMPLE_END = pd.Period("2002-01", "M"), pd.Period("2026-07", "M")


def download_prices(tickers, start, end):
    """Monthly adjusted closing prices from Yahoo Finance."""
    install_and_import("yfinance", "yf")
    raw = yf.download(tickers, start=start, end=end, interval="1mo",
                      auto_adjust=False, actions=False, progress=False)
    px = raw["Adj Close"][tickers].dropna(how="all")
    px.index = pd.to_datetime(px.index).to_period("M")          # 2026-07-01 -> 2026-07
    px.index.name = "date"
    return px


def download_french(zip_name):
    """Monthly table (in percent) from a zipped CSV file of the Kenneth French data library."""
    import io, zipfile, requests
    resp = requests.get(FRENCH_URL + zip_name, timeout=60)
    resp.raise_for_status()
    with zipfile.ZipFile(io.BytesIO(resp.content)) as z:
        text = z.read(z.namelist()[0]).decode("latin-1")
    header, rows = None, []
    for line in text.splitlines():
        parts = [s.strip() for s in line.split(",")]
        if header is None:                                       # first line like ",Mkt-RF,SMB,HML,RF"
            if len(parts) > 1 and parts[0] == "" and parts[1] != "":
                header = parts[1:]
            continue
        if len(parts) > 1 and parts[0].isdigit() and len(parts[0]) == 6:
            rows.append(parts)                                   # monthly row, e.g. "200201,-1.44,..."
        elif rows:                                               # end of the monthly block (annual data follow)
            break
    df = pd.DataFrame([r[1:] for r in rows], index=[r[0] for r in rows], columns=header).astype(float)
    df.index.name = "date"
    return df


# ---- fund prices ----
if PRICE_FILE.exists():
    prices = pd.read_csv(PRICE_FILE, index_col="date")
    prices.index = pd.PeriodIndex(prices.index, freq="M")
    print(f"Fund prices loaded from local copy: {PRICE_FILE.name}")
else:
    prices = download_prices(FUNDS, PRICE_START, PRICE_END)
    prices.to_csv(PRICE_FILE, index_label="date")
    print("Fund prices downloaded from Yahoo Finance and saved.")

# ---- factors (in percent, as in the French library) ----
if FACTOR_FILE.exists():
    factors_pct = pd.read_csv(FACTOR_FILE, index_col="date")
    print(f"Factors loaded from local copy: {FACTOR_FILE.name}")
else:
    ff3 = download_french("F-F_Research_Data_Factors_CSV.zip")
    mom = download_french("F-F_Momentum_Factor_CSV.zip")
    factors_pct = ff3.join(mom, how="inner")
    factors_pct.to_csv(FACTOR_FILE, index_label="date")
    print("Factors downloaded from the Kenneth French data library and saved.")

factors_pct.index = pd.to_datetime(factors_pct.index.astype(str), format="%Y%m").to_period("M")
factors_pct.columns = [c.strip() for c in factors_pct.columns]

prices = prices.loc[SAMPLE_START - 1:SAMPLE_END, FUNDS]          # December 2001 to July 2026
factors = factors_pct.loc[SAMPLE_START:SAMPLE_END, ["Mkt-RF", "SMB", "HML", "Mom", "RF"]] / 100
prices.tail(3)
```

```python
rets = prices.pct_change().dropna()                              # simple monthly total returns
assert rets.index.equals(factors.index), "Fund and factor months do not line up"
assert rets.notna().all().all() and factors.notna().all().all(), "Missing observations"

rf = factors["RF"]
excess = rets.sub(rf, axis=0)                                    # fund excess returns
mkt = factors["Mkt-RF"]                                          # market excess return

print(f"Sample: {excess.index[0].strftime('%b %Y')} to {excess.index[-1].strftime('%b %Y')}, T = {len(excess)} months "
      f"(identical for all four funds and the factors)")
excess.head()
```

For comparison we also report the annualised geometric total return, $\left(\prod_t (1+r_{i,t})\right)^{12/T}-1$, i.e. the realised compound growth rate of an investment in the fund. The market's total return is $(r_{M,t}-r_{f,t}) + r_{f,t}$.

## Sharpe ratio and information ratio

The Sharpe ratio measures the expected excess return per unit of total risk,
$$
\hat S_i = \frac{\bar r^e_i}{\hat\sigma_i},
$$
where $\bar r^e_i$ and $\hat\sigma_i$ are the sample mean and standard deviation of the monthly excess returns. It is the slope of the CAL obtained by combining the fund with T-bills, so it is the relevant measure if the fund is an investor's *entire* risky portfolio. With i.i.d. returns the mean scales with 12 and the standard deviation with $\sqrt{12}$, so $\hat S^{ann}_i = \sqrt{12}\,\hat S_i$.

The information ratio measures performance relative to a benchmark, here the market. The active return is the difference between the fund return and the benchmark return, in which the risk-free rate cancels out,
$$
a_{i,t} = r_{i,t} - r_{M,t} = r^e_{i,t} - r^e_{M,t},
\qquad
\widehat{IR}_i = \frac{\bar a_i}{sd(a_i)},
$$
where the denominator $sd(a_i)$ is the **tracking error**. The IR is the average outperformance relative to the market per unit of risk taken by deviating from it, and is annualised with $\sqrt{12}$ as well. Since $sd(a_i)/\sqrt{T}$ is the standard error of the mean active return, the t-statistic for $H_0: \mathbb{E}[a_{i}] = 0$ (no out- or underperformance of the market) is simply the monthly information ratio multiplied by $\sqrt{T}$,
$$
t_{\bar a_i} = \frac{\bar a_i}{sd(a_i)/\sqrt{T}} = \widehat{IR}_i\,\sqrt{T} .
$$

**Appraisal ratio (CAPM).** The IR implicitly assumes that the fund has a market beta of one. As a complement, the regression-based appraisal ratio first removes the fund's market exposure,
$$
r^e_{i,t} = \alpha_i + \beta_i\, r^e_{M,t} + \varepsilon_{i,t},
\qquad
\widehat{AR}_i = \frac{\hat\alpha_i}{sd(\hat\varepsilon_i)} .
$$
Averaging the regression over the sample (OLS residuals have mean zero) links the two measures exactly:
$$
\bar a_i = \bar r^e_i - \bar r^e_M = \hat\alpha_i + (\hat\beta_i - 1)\,\bar r^e_M .
$$
The average active return thus consists of CAPM alpha plus a part that only reflects a market exposure above or below one.

```python
active = excess.sub(mkt, axis=0)                                 # a_t = r_i - r_M
t_active = active.mean() / (active.std() / np.sqrt(len(active)))   # = monthly IR x sqrt(T)
p_active = pd.Series(2 * stats.t.sf(t_active.abs(), df=len(active) - 1), index=FUNDS)
capm = {k: sm.OLS(excess[k], sm.add_constant(mkt)).fit() for k in FUNDS}

perf = pd.DataFrame({
    "Mean excess return (ann.)": excess.mean() * 12,
    "SD (ann.)": excess.std() * np.sqrt(12),
    "Sharpe ratio (monthly)": excess.mean() / excess.std(),
    "Sharpe ratio (ann.)": excess.mean() / excess.std() * np.sqrt(12),
    "Mean active return (ann.)": active.mean() * 12,
    "Tracking error (ann.)": active.std() * np.sqrt(12),
    "Information ratio (monthly)": active.mean() / active.std(),
    "Information ratio (ann.)": active.mean() / active.std() * np.sqrt(12),
    "t-stat mean active return": t_active,
    "p-value mean active return": p_active,
    "CAPM beta": pd.Series({k: capm[k].params["Mkt-RF"] for k in FUNDS}),
    "CAPM R-squared": pd.Series({k: capm[k].rsquared for k in FUNDS}),
    "CAPM alpha (ann.)": pd.Series({k: capm[k].params["const"] * 12 for k in FUNDS}),
    "Beta effect, (beta - 1) x mean mkt excess (ann.)": pd.Series({k: (capm[k].params["Mkt-RF"] - 1) * mkt.mean() * 12 for k in FUNDS}),
    "Appraisal ratio (ann.)": pd.Series({k: capm[k].params["const"] / np.sqrt(capm[k].mse_resid) * np.sqrt(12) for k in FUNDS}),
})
perf.loc["MKT"] = [mkt.mean() * 12, mkt.std() * np.sqrt(12), mkt.mean() / mkt.std(), mkt.mean() / mkt.std() * np.sqrt(12),
                   0, 0, np.nan, np.nan, np.nan, np.nan, 1, 1, 0, 0, np.nan]

pct_cols = ["Mean excess return (ann.)", "SD (ann.)", "Mean active return (ann.)", "Tracking error (ann.)",
            "CAPM alpha (ann.)", "Beta effect, (beta - 1) x mean mkt excess (ann.)"]
perf.rename(index=NAMES).T.style.format(
    lambda v: "" if pd.isna(v) else f"{v:.3f}"
).format(lambda v: f"{v:.2%}", subset=pd.IndexSlice[pct_cols, :])
```

**Sharpe ratios:** None of the four funds achieved a higher Sharpe ratio than the market (0.610). TRBCX (0.575) and STSEX (0.574) come closest, followed by GLCGX (0.522) and DFSVX (0.499). It is interesting that TRBCX and STSEX reach almost the same Sharpe ratio in very different ways: TRBCX earns a high average excess return (9.85%) with high volatility (17.1%), whereas STSEX earns less (7.61%) with much lower volatility (13.3%). **Information ratios:** Relative to the market, TRBCX (IR = 0.089) and DFSVX (IR = 0.124) outperformed on average by 0.54% and 1.40% p.a., with tracking errors of 6.1% and 11.3%. GLCGX (IR = -0.173) and STSEX (IR = -0.288) underperformed by 0.78% and 1.70% p.a. All information ratios are small. **Evidence of skill:** The decomposition $\bar a_i = \hat\alpha_i + (\hat\beta_i - 1)\bar r^e_M$ shows that most of the active returns are explained by market beta and not by alpha. Conversely, STSEX's underperformance of 1.70% p.a. comes entirely from its low beta of 0.80 (-1.83% p.a.), while its CAPM alpha is slightly positive (+0.14%). In a period in which the market earned a high premium (9.31% p.a.), any fund with $\beta > 1$ looks good and any fund with $\beta < 1$ looks bad against the market, even though an investor could create the same market exposure with leverage or T-bills at no cost. The appraisal ratios, which correct for beta, are all close to zero (between -0.23 for GLCGX and +0.03 for STSEX). Whether the remaining small alphas reflect skill or only exposure to other systematic factors such as size and value is what we test in the next section.

## Evidence of fund manager skill: factor regressions

A manager has skill if the fund earns a higher return than an investor could obtain with a passive portfolio that has the same systematic risk exposures. In a factor model, this abnormal return is measured by the intercept $\alpha_i$ of a time-series regression of the fund's excess return on the factor returns. Since the right-hand side variables are returns on traded (excess-return) portfolios, $\sum_k \beta_{i,k} f_{k,t}$ is the return of a passive replicating portfolio of the factors and T-bills, and $\alpha_i$ is the average return of the fund in excess of that replicating portfolio. If the factor model correctly describes expected returns, $\alpha_i = 0$ for a fund without skill.

**Fama-French three-factor model (FF3):**
$$
r^e_{i,t} = \alpha_i + \beta_{i,M}\,(r_{M,t}-r_{f,t}) + \beta_{i,S}\,SMB_t + \beta_{i,H}\,HML_t + \varepsilon_{i,t}
$$

**Fama-French-Carhart four-factor model (FFC4):**
$$
r^e_{i,t} = \alpha_i + \beta_{i,M}\,(r_{M,t}-r_{f,t}) + \beta_{i,S}\,SMB_t + \beta_{i,H}\,HML_t + \beta_{i,Mom}\,Mom_t + \varepsilon_{i,t}
$$

The factor loadings describe the fund's investment style: $\beta_{i,S} > 0$ indicates a tilt towards small stocks, $\beta_{i,H} > 0$ a tilt towards value stocks and $\beta_{i,H} < 0$ towards growth stocks, $\beta_{i,Mom} > 0$ a tilt towards past winners. A manager who simply follows a size, value or momentum style earns the corresponding factor premia, but this is not skill in the sense of the models because the same exposure can be obtained passively.

The following details the hypothesis test. A positive estimated alpha alone is not enough, since it could be the result of luck. For each fund we test
$$
H_0: \alpha_i = 0 \quad \text{against} \quad H_1: \alpha_i \neq 0,
\qquad
t_{\alpha_i} = \frac{\hat\alpha_i}{SE(\hat\alpha_i)},
$$
and reject $H_0$ at the conventional two-sided 5% level if $|t| > 1.97$ ($p < 0.05$). Evidence of skill requires a significantly *positive* alpha. Our main results use conventional OLS standard errors. Because monthly fund returns may show heteroskedasticity and some autocorrelation, we also report the p-value based on Newey-West (HAC) standard errors with 6 lags as a robustness check. The usual $4(T/100)^{2/9}$ rule of thumb gives five lags for $T = 295$, so this choice is slightly conservative. Alphas are monthly in the regressions and annualised by multiplying by 12. To see how alpha changes as factors are added, we also estimate the CAPM regression from Section 2 as a baseline.

```python
MODELS = {"CAPM": ["Mkt-RF"], "FF3": ["Mkt-RF", "SMB", "HML"], "FFC4": ["Mkt-RF", "SMB", "HML", "Mom"]}

res, res_hac = {}, {}
for m, cols in MODELS.items():
    X = sm.add_constant(factors[cols])
    for k in FUNDS:
        res[m, k] = sm.OLS(excess[k], X).fit()
        res_hac[m, k] = sm.OLS(excess[k], X).fit(cov_type="HAC", cov_kwds={"maxlags": 6})


def alpha_table(model):
    rows = {}
    for k in FUNDS:
        r, h = res[model, k], res_hac[model, k]
        lo, hi = r.conf_int().loc["const"] * 12
        rows[k] = {
            "alpha (monthly)": r.params["const"],
            "alpha (ann.)": r.params["const"] * 12,
            "SE alpha (ann.)": r.bse["const"] * 12,
            "95% CI alpha (ann.)": f"[{lo:.2%}, {hi:.2%}]",
            "t(alpha)": r.tvalues["const"],
            "p-value": r.pvalues["const"],
            "p-value (Newey-West)": h.pvalues["const"],
            **{f"beta {c}": r.params[c] for c in MODELS[model]},
            **{f"t({c})": r.tvalues[c] for c in MODELS[model]},
            "R-squared": r.rsquared,
            "Adj. R-squared": r.rsquared_adj,
            "Residual SD (ann.)": np.sqrt(r.mse_resid * 12),
            "Appraisal ratio (ann.)": r.params["const"] / np.sqrt(r.mse_resid) * np.sqrt(12),
        }
    return pd.DataFrame(rows)


def show(tab):
    pct = ["alpha (ann.)", "SE alpha (ann.)", "Residual SD (ann.)"]
    return tab.rename(columns=NAMES).style.format(
        lambda v: v if isinstance(v, str) else f"{v:.4f}"
    ).format(lambda v: f"{v:.2%}", subset=pd.IndexSlice[pct, :]
    ).format(lambda v: f"{v:.3%}", subset=pd.IndexSlice[["alpha (monthly)"], :])


def regression_output(model):
    out = summary_col([res[model, k] for k in FUNDS], model_names=FUNDS, stars=True,
                      float_format="%.5f", info_dict={"N": lambda r: f"{int(r.nobs)}"},
                      regressor_order=["const"] + MODELS[model])
    print(f"{model} regressions of monthly fund excess returns (standard errors in parentheses)")
    print(out)
```

```python
halves = {"2002-01 to 2013-12": (pd.Period("2002-01", "M"), pd.Period("2013-12", "M")),
          "2014-01 to 2026-07": (pd.Period("2014-01", "M"), pd.Period("2026-07", "M"))}
sub = {}
for lab, (s, e) in halves.items():
    X = sm.add_constant(factors.loc[s:e, MODELS["FFC4"]])
    for k in FUNDS:
        r = sm.OLS(excess.loc[s:e, k], X).fit()
        sub[k] = sub.get(k, {}) | {(lab, "alpha (ann.)"): r.params["const"] * 12,
                                   (lab, "t(alpha)"): r.tvalues["const"],
                                   (lab, "p-value"): r.pvalues["const"]}
sub = pd.DataFrame(sub).T
sub.columns = pd.MultiIndex.from_tuples(sub.columns, names=["FFC4, subperiod", ""])
sub.rename(index=NAMES).style.format("{:.3f}").format(
    "{:.2%}", subset=[c for c in sub.columns if c[1] == "alpha (ann.)"])
```

### Conclusion: is there evidence of fund manager skill?

No, neither the FF3 nor the FFC4 model provides evidence of manager skill for any of the four funds. All estimated alphas are statistically indistinguishable from zero at the 5% level, three of them are within $\pm 0.15\%$ p.a. of zero under both models, and the only economically sizeable alpha (GLCGX, about -1.1% to -1.3% p.a.) is negative and marginally significant at the 10% level under FFC4. The results are the same with Newey-West standard errors and in both halves of the sample, apart from GLCGX's significant underperformance before 2014.

The funds' performance is instead almost entirely explained by their exposures to the market, size, value and momentum factors. This is consistent with the results from Section 2. The positive information ratios of TRBCX and DFSVX are the result of market betas above one and, for DFSVX, of its small-cap value style, not of superior stock selection. An investor could have replicated the risk and return of each fund with a passive portfolio of the factors, which is precisely the case when $\alpha = 0$.

There are several points to consider when interpreting these results:

- Net versus gross alpha: The fund returns are net of fees. An alpha of zero after fees implies a small positive alpha *before* fees, roughly equal to the annual expense ratio of the fund (a fraction of a percent per year for these funds). The managers may therefore have some ability to generate returns, but it is just enough to cover their fees, so that investors do not benefit from it. This is consistent with the "arithmetic of active management" (Sharpe, 1991): since passive investors hold the market portfolio, active investors in aggregate must hold it as well, so the average active fund earns a gross alpha of about zero and a negative net alpha after fees. It is also consistent with Berk and Green (2004), where fund flows compete away any net alpha.
- Statistical power: The standard errors of the annualised alphas are between 0.7% and 0.9%, so the 95% confidence intervals are about $\pm 1.3\%$ to $\pm 1.8\%$ p.a. wide. Even with almost 25 years of data, a true alpha of around +1% p.a., which would be economically very valuable, could not be distinguished from zero. The absence of evidence of skill is therefore not the same as evidence of the absence of skill, although the point estimates are centred close to zero.
- The factor benchmark is not free: SMB, HML and Mom are returns on long-short paper portfolios that ignore trading costs, short-selling costs and fees. For DFSVX, an alpha of zero relative to these factors means that the fund delivered its size and value exposure without losing return to costs, which still has value for investors who want this exposure. Measured against investable index funds instead of paper portfolios, its performance would probably look better.
- Selection and multiple testing: The four funds were chosen ex post among funds that survived the full 25-year period, which biases the sample towards successful funds. Moreover, with four funds and three model specifications, finding one alpha that is significant at the 10% level is not unusual even if all true alphas were zero. Both effects make it easier, not harder, to find positive alphas by chance, which strengthens the conclusion that there is no evidence of skill.

All in all, the regressions suggest that what investors pay for in these funds is not stock-picking skill but systematic factor exposure: market risk for all four funds, a growth tilt for TRBCX and GLCGX, a large-cap value tilt for STSEX and a small-cap value tilt for DFSVX.
