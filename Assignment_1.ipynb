## Expected return, standard deviation and correlation coefficient

The expected return $\mathbb{E}[r]$ is given by
$$
\mathbb{E}[r] = \sum_{s=1}^{S} p_s r_s ,
$$
where $S$ is the total number of states and $p_s$ is the probability of state $s$.
Since all three states are equally likely, $p_s = \frac{1}{S}$ for all $s$, this simplifies to
$$
\mathbb{E}[r] = \frac{1}{S} \sum_{s=1}^S r_s .
$$

Substituting the values from the table yields

$$
\mathbb{E}[R_A] = \tfrac{1}{3}\big(-2 - 1 + 3\big) = 0\% 
$$

$$
\mathbb{E}[R_B] = \tfrac{1}{3}\big(5 + 3 - 5\big) = 1\% .
$$

---

The variance of returns is given by
$$
\mathrm{Var}(r) = \sigma^2 = \sum_{s=1}^{S} p_s \big(r_s - \mathbb{E}[r]\big)^2 ,
$$
with the  standard deviation being its square root
$$
\sigma = \sqrt{\mathrm{Var}(r)} = \sqrt{\sum_{s=1}^{S} p_s \big(r_s - \mathbb{E}[r]\big)^2} .
$$
As above, $p_s$ can be pulled out of the sum as the common factor.

Substituting the values from the table yields

$$
\begin{aligned}
\sigma_A    &= \sqrt{\tfrac{1}{3}\big[(-2-0)^2 + (-1-0)^2 + (3-0)^2\big]}
            = \sqrt{\tfrac{14}{3}} \approx 2.16\%, \\[6pt]
\sigma_B    &= \sqrt{\tfrac{1}{3}\big[(5-1)^2 + (3-1)^2 + (-5-1)^2\big]}
            = \sqrt{\tfrac{56}{3}} \approx 4.32\% .
\end{aligned}
$$

---

The correlation coefficient is given by
$$
\rho_{ij} = \mathrm{Corr}[r_i, r_j] = \rho_{ij} = \frac{\sigma_{ij}}{\sigma_i \sigma_j} ,
$$
where the covariance $\mathrm{Cov}[r_i, r_j]$ is defined as
$$
\mathrm{Cov}[r_i, r_j] = \sigma_{ij} = \mathbb{E}\big[(r_i - \mathbb{E}[r_i])(r_j - \mathbb{E}[r_j])\big] .
$$

Writing $\mu_i = \mathbb{E}[r_i]$ and $\mu_j = \mathbb{E}[r_j]$, which are constants,
expanding the product and applying linearity of expectations gives

$$
\begin{aligned}
\sigma_{ij} &= \mathbb{E}\big[r_i r_j - \mu_j r_i - \mu_i r_j + \mu_i \mu_j\big] \\[4pt]
            &= \mathbb{E}[r_i r_j] - \mu_j \mathbb{E}[r_i] - \mu_i \mathbb{E}[r_j] + \mu_i \mu_j \\[4pt]
            &= \mathbb{E}[r_i r_j] - \mu_i \mu_j \\[4pt]
            &= \mathbb{E}[r_i r_j] - \mathbb{E}[r_i]\,\mathbb{E}[r_j] .
\end{aligned}
$$

Substituting the values from the table yields

$$
\sigma_{AB} = \tfrac{1}{3}\big[(-2)(5) + (-1)(3) + (3)(-5)\big] - (0)(1)
= -\tfrac{28}{3} \approx -9.33 \ (\%^2)
$$

$$
\rho_{AB} = \frac{-\tfrac{28}{3}}{\sqrt{\tfrac{14}{3}} \cdot \sqrt{\tfrac{56}{3}}}
= \frac{-\tfrac{28}{3}}{\sqrt{\tfrac{784}{9}}}
= \frac{-\tfrac{28}{3}}{\tfrac{28}{3}}
= -1 .
$$
The assets are perfectly negatively correlated.

## The fully invested portfolio, made of A and B, with minimum variance

Since the portfolio is fully invested it follows that $w_B = 1 - w_A$ and the portfolio variance can be written as

$$
\sigma_p^2 = w_A^2 \sigma_A^2 + (1-w_A)^2 \sigma_B^2 + 2 w_A (1-w_A) \sigma_{AB} .
$$

To find the minimum Variance Portfolio we differentiate the variance function with respect to $w_A$,

$$
\frac{\partial \sigma_p^2}{\partial w_A}
= 2 w_A \sigma_A^2 - 2(1-w_A)\sigma_B^2 + (2 - 4 w_A)\sigma_{AB} ,
$$

and set the derivative equal to zero

$$
\begin{aligned}
0 &= 2 w_A \sigma_A^2 - 2(1-w_A)\sigma_B^2 + (2-4w_A)\sigma_{AB} \\
0 &= w_A \sigma_A^2 - \sigma_B^2 + w_A \sigma_B^2 + \sigma_{AB} - 2 w_A \sigma_{AB} \\
0 &= w_A\big(\sigma_A^2 + \sigma_B^2 - 2\sigma_{AB}\big) - \sigma_B^2 + \sigma_{AB} \\
w_A^{*} &= \frac{\sigma_B^2 - \sigma_{AB}}{\sigma_A^2 + \sigma_B^2 - 2\sigma_{AB}} .
\end{aligned}
$$

Substituting $\sigma_A^2 = \tfrac{14}{3}$, $\sigma_B^2 = \tfrac{56}{3}$ and $\sigma_{AB} = -\tfrac{28}{3}$

$$
w_A^{*} = \frac{\tfrac{56}{3} + \tfrac{28}{3}}{\tfrac{14}{3} + \tfrac{56}{3} + \tfrac{56}{3}}
= \frac{28}{42} = \frac{2}{3},
\qquad w_B^{*} = \frac{1}{3} .
$$

The resulting portfolio variance is

$$
\sigma_p^2 = \tfrac{4}{9}\cdot\tfrac{14}{3} + \tfrac{1}{9}\cdot\tfrac{56}{3}
+ 2\cdot\tfrac{2}{3}\cdot\tfrac{1}{3}\cdot\left(-\tfrac{28}{3}\right)
= \tfrac{56}{27} + \tfrac{56}{27} - \tfrac{112}{27} = 0 .
$$

## The set of all possible arbitrage portfolios, when a risk-free bond with return 0% exists and a no-short sale constraint is imposed on A and B

The zero-cost portfolio therefore borrows at $r_f = 0\%$ and invests the money in the two risky assets. To begin with, we normalise the borrowed amount to one unit and 

$$
w_A + w_B = 1, \qquad w_A, w_B \geq 0 ,
$$

so that $w_B = 1 - w_A$ with $w_A \in [0,1]$ .  

As the bonds return is 0, the portfolio return in state $s$ is $r_{P, s} =  w_A r_{A,s} + (1-w_A) r_{B,s}$. Following the no-arbitrage condition this must yield a non-negative amount in every state, which allows us to set up three inequalities

$$
\begin{aligned}
\text{State 1:} \quad 0 &\leq -2 w_A + 5(1-w_A) = 5 - 7 w_A
  && \Longrightarrow \quad w_A \leq \tfrac{5}{7} \approx 0.7143 , \\[4pt]
\text{State 2:} \quad 0 &\leq -1 w_A + 3(1-w_A) = 3 - 4 w_A
  && \Longrightarrow \quad w_A \leq \tfrac{3}{4} = 0.75 , \\[4pt]
\text{State 3:} \quad 0 &\leq \phantom{-}3 w_A - 5(1-w_A) = 8 w_A - 5
  && \Longrightarrow \quad w_A \geq \tfrac{5}{8} = 0.625 .
\end{aligned}
$$

Since the first condition is stricter than the second one, only conditions (1) and (3) are binding.
Combining the remaining two yields

$$
\tfrac{5}{8} \leq w_A \leq \tfrac{5}{7}
\qquad\text{i.e.}\qquad 0.625 \leq w_A \leq 0.7143 .
$$

As their exists no $w_A$ which satisfies all three conditions with equality
simultaneously, i.e. $w_A = \tfrac{5}{7}$, $w_A = \tfrac{3}{4}$ and $w_A = \tfrac{5}{8}$, each portfolio earns a strictly positive return in atleast one state.

By normalising to a single unit of borrowed money, we obtained the proportions of $w_A$ and $w_B$. Scaling the position up or down by shorting more or less of the risk-free asset gives all possible arbitrage portfolios $(w_A, w_B, w_f)$ as $x \cdot (w_A,\ 1-w_A,\ -1)$ for any $x > 0$ and any $w_A \in \left[\tfrac{5}{8}, \tfrac{5}{7}\right]$.

## Compute and plot the efficient frontier

The solve this we use `opt_portfolio_2asset.ipynb` with the values referring to the table above.

```python
import numpy as np
import matplotlib.pyplot as plt

# Setup
# Basic inputs for bonds and stocks

# Mean
SB_mean = np.array([[0.1], [0.05]])

# Variance-Covariance Matrix
SB_Sigma = np.array([[0.2**2, 0.0002], [0.0002, 0.08**2]])

# Vector of ones
ones2 = np.ones((2, 1))

# Inverse of the variance-covariance matrix
SB_invSigma = np.linalg.inv(SB_Sigma)
```

```python
# Define vectors for weights: Stock = (0,1)^t and Bond = (1,0)^t
E_w = np.array([[1], [0]])  # Bond
B_w = np.array([[0], [1]])  # Stock

# Create many portfolios as a combination of stock and bond

# Possible combinations of weights
y = np.arange(-1, 2.01, 0.01)

# Convert y to a 2D array where each row is a weight for a portfolio
y = y.reshape(1, -1)
port = np.dot(B_w, y) + np.dot(E_w, (1 - y))

# Compute expected returns
port_mean = np.dot(port.T, SB_mean)

# Compute var-covar matrix of portfolio returns (we will only use the variances on the main diagonal)
port_var = np.diag(np.dot(np.dot(port.T, SB_Sigma), port))
port_var = np.clip(port_var, 0, None)  # guard against tiny negative values from float rounding
port_vol = np.sqrt(port_var).reshape(len(port_var), 1)
```

```python
# Plot MV
plt.figure(figsize=(10, 6))
plt.plot(port_vol * 100, port_mean * 100, 'b-', label='Efficient Frontier')
plt.title("Mean vs. SD")
plt.xlabel("Standard Deviation (%)")
plt.ylabel("Mean Return (%)")
plt.xlim(0, 25)
plt.ylim(0, 16)

# Add points for stock and bond funds
plt.scatter(np.sqrt(np.diag(SB_Sigma)) * 100, SB_mean * 100, color='red')
SB_labels = ["Stocks", "Bonds"]

# Annotate points
for i, txt in enumerate(SB_labels):
    plt.annotate(txt, (np.sqrt(np.diag(SB_Sigma))[i] * 100, SB_mean[i, 0] * 100), textcoords="offset points", xytext=(0, 5), ha='center')

plt.legend()
plt.grid(True)
plt.show()
```

## Compute the asset weights that make up the minimum variance portfolio (GMV)

Following *Section 1.3* the weights of the fully invested global minimum variance portfolio, with 2 risky assets, are
$$
w_A^{*} = \frac{\sigma_B^2 - \sigma_{AB}}{\sigma_A^2 + \sigma_B^2 - 2\sigma_{AB}} \qquad \text{and} \qquad w_B^{*} = 1-w_A^{*} .
$$

Substituting $\sigma_A^2 = 0.2^2$, $\sigma_B^2 = 0.08^2$ and $\sigma_{AB} = 0.0002$

$$
w_A^{*} = \frac{0.08^2 - 0.0002}{0.2^2 + 0.08^2 - 0.0004}
= \frac{31}{230}\approx 0.1349,
\qquad w_B^{*} = \frac{199}{230} \approx 0.8652 .
$$

The resulting portfolio variance is
$$
\sigma_P^2 = \left(\tfrac{31}{230}\right)^2 0.04 + \left(\tfrac{199}{230}\right)^2 0.0064
+ 2\cdot\tfrac{31}{230}\cdot\tfrac{199}{230}\cdot 0.0002 \approx 0.005564 ,
$$

and the standard deviation is
$$
\sigma_P \approx 0.074595 = 7.46\% .
$$

The expected return of the GMV portfolio is

\mu_P = \left(\tfrac{31}{230}\right) 0.10 + \left(\tfrac{199}{230}\right) 0.05 \approx 0.0567 = 5.67\% .
$$

It can be seen that holding a certain amount of the riskier stock fund reduces the overall portfolio risk compared to the safer bond fund,  $\sigma_P = 7.46\% < \sigma_B = 8\%$.

## Derivation of the tangent portfolio and the optimal CAL, with $r_f = 0.03$

The tangent portfolio maximizes the Sharpe Ratio $S$, which is defined as 

$$
S = \frac{\mathbb{E}[r_T]-r_f}{\sigma_T} .
$$

For the case with two risky assets let $g(w_A)$ be the excess return function and $h(w_A)$ be the variance function
$$
\begin{aligned}
g &= w_A r_A + (1-w_A) r_B - r_f \\
h &= w_A^2\sigma_A^2 + (1-w_A)^2\sigma_B^2 + 2w_A(1-w_A)\sigma_{AB} .
\end{aligned}
$$
As the tangency portfolio lies on the efficient frontier and is as such made up of only the two risky assets, we can use $w_B = 1 - w_A$.

To maximize the Sharpe Ratio we differentiate $S$ with respect to $w_A$ and set this FOC equal to 0
$$
\begin{aligned}
S(w_A) &= \frac{g}{\sqrt{h}} \\
S(w_A) &= g\,h^{-1/2} .
\end{aligned}
$$

By applying the product and chain rule we find the FOC
$$
\begin{aligned}
\frac{\partial S}{\partial w_A}     &= g'h^{-1/2} + g\left(-\tfrac{1}{2}\right)h^{-3/2}h' \\
                                    &= h^{-3/2}\left[g'h - \tfrac{1}{2}gh'\right] .
\end{aligned}
$$

Now we can set the derivative equal to 0 and multiply by $h^{3/2}$, which gives
$$
\begin{aligned}
g'h - \tfrac{1}{2}gh' &= 0 \\
2g'h                  &= gh' .
\end{aligned}
$$

Substituting into $2g'h = gh'$, differentiating the remaining parts and simpliyfing, results in
$$
w_A^{T} = \frac{(r_A-r_f)\sigma_B^2-(r_B-r_f)\sigma_{AB}} {(r_A-r_f)\sigma_B^2+(r_B-r_f)\sigma_A^2-\big[(r_A-r_f)+(r_B-r_f)\big]\sigma_{AB}} .
$$

Substituting $r_A = 0.10$, $r_B = 0.05$, $r_f = 0.03$, $\sigma_A^2 = 0.2^2$, $\sigma_B^2 = 0.08^2$ and $\sigma_{AB} = 0.0002$
$$
w_A^{T} = \frac{(0.07)0.0064 - (0.02)0.0002}{(0.07)0.0064 + (0.02)0.04 - (0.09)0.0002} = \frac{74}{205} \approx 0.3610, \qquad w_B^{T} = \frac{131}{205} \approx 0.6390 .
$$

The optimal CAL is the linear function through the risk-free asset and the tangency portfolio $T$ with $r_f$ as the y-intercept and the Sharpe ratio as the slope

$$
\mathrm{CAL} = r_f + \frac{\mathbb{E}[r_T]-r_f}{\sigma_T}\,\sigma_{CAL}
= 0.03 + 0.4276\,\sigma_C .
$$

---

The following code was adapted from `opt_portfolio_2asset.ipynb` with the values referring to the results from above.

```python
# Risk-free rate
rf = 0.03

# Excess return
SB_excess = SB_mean - rf

# Tangent portfolio weights
wT1 = (SB_Sigma[1, 1] * SB_excess[0] - SB_Sigma[1, 0] * SB_excess[1]) / \
      (SB_Sigma[0, 0] * SB_excess[1] + SB_Sigma[1, 1] * SB_excess[0] - SB_Sigma[0, 1] * np.sum(SB_excess))

# Ensure wT1 is a scalar
wT1 = wT1[0]

T_w = np.array([[wT1], [1 - wT1]])

# Mean and volatility of tangent portfolio (as scalars)
T_mean = np.dot(T_w.T, SB_mean).item()
T_vol = np.sqrt(np.dot(np.dot(T_w.T, SB_Sigma), T_w).item())

# Sharpe ratio
T_SR = (T_mean - rf) / T_vol

print(f"w_A = {wT1:.4f},  w_B = {1 - wT1:.4f}")
print(f"E[r] = {T_mean*100:.3f}%,  sd = {T_vol*100:.3f}%,  Sharpe = {T_SR:.4f}")

# Possible combinations of weights
yT = np.arange(-10, 10.01, 0.01)
yT = yT.reshape(1, -1)

# Create many portfolios using E and B
portT = np.dot(E_w, yT) + np.dot(B_w, (1 - yT))

# Compute expected returns
portT_mean = np.dot(portT.T, SB_mean)

# Compute var-covar matrix of portfolio returns (we will only use the variances on the main diagonal)
portT_var = np.diag(np.dot(np.dot(portT.T, SB_Sigma), portT))
portT_vol = np.sqrt(portT_var).reshape(len(portT_var), 1)

# Compute var-covar matrix of portfolio returns (we will only use the variances on the main diagonal)
portT_var = np.diag(np.dot(np.dot(portT.T, SB_Sigma), portT))
portT_vol = np.sqrt(portT_var).reshape(len(portT_var), 1)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(portT_vol * 100, portT_mean * 100, 'b-', label='Efficient Frontier')
plt.title("Mean vs. SD")
plt.xlabel("Standard Deviation (%)")
plt.ylabel("Mean Return (%)")
plt.xlim(0, 25)
plt.ylim(0, 16)

# Add points for Tangent portfolio
plt.scatter(T_vol * 100, T_mean * 100, color='green', marker='o')
plt.text(T_vol * 100, T_mean * 100, 'Tangent Portfolio', fontsize=9, ha='center', va='bottom')

# Add points for stock and bond funds
plt.scatter(np.sqrt(np.diag(SB_Sigma)) * 100, SB_mean * 100, color='red', marker='o')
SB_labels = ["Stocks", "Bonds"]
for i, txt in enumerate(SB_labels):
    plt.text(np.sqrt(np.diag(SB_Sigma))[i] * 100, SB_mean[i, 0] * 100, txt,
             fontsize=9, ha='center', va='bottom')

# Add the CAL to the picture

# Possible combinations of weights
yf = np.arange(0, 10.01, 0.01)

# Compute expected returns
CAL_mean = rf + yf * (T_mean - rf)

# Compute volatilities
CAL_vol = yf * T_vol

# Add CAL to the graph
plt.plot(CAL_vol * 100, CAL_mean * 100, color='cyan', label='Capital Allocation Line (CAL)')

plt.legend()
plt.grid(True)
plt.show()
```

## The optimal portfolio for an investor with risk aversion parameter a = 4 and its certainity equivalent

The investor can now move along the CAL by splitting his investment between the risky tangency portfolio and the risk-free asset. Therefore, a risk averse investor maximizes his utility according to the following utility function
$$
U = r_f + w^{*}(\mathbb{E}[r_T]-r_f)-\frac{a}{2}(w^*)^2\sigma_T^2 .
$$

To find the maximum we differentiate with respect to $w$ and set it equal to 0

$$
\frac{dU}{dw} = (\mathbb{E}[r_T]-r_f) - a\,w\,\sigma_T^2 = 0 \quad \Longrightarrow \quad
w^* = \frac{(\mathbb{E}[r_T]-r_f)}{a\,\sigma_T^2} .
$$

Substituting the values yields
$$
w^* \approx \frac{0.06805 - 0.03}{4(0.007918)} = 1.2014 .
$$
To maximize his utility the investor takes on leverage by shortselling 20.14\% of the risk-free asset and investing 120.14\% into the tangency portfolio.

---

The certainty equivalent is the risk-free rate $r_f^*$ that provides the investor with the same utility as the above portfolio. As a sole investment in a risk-free asset carries no risk (volatilty), the utility functions simplifies to
$$
U = r_f^* .
$$

To find this rate we set the 'risk-free utility function' equal to the general utility function from above
$$
\begin{aligned}
r_f^* &= r_f + w^{*}(\mathbb{E}[r_T]-r_f)-\frac{a}{2}(w^*)^2\sigma_T^2 \\
r_f^* &\approx 0.03 + 1.2014(0.06805 - 0.03)-\frac{4}{2}(1.2014)^2(0.007918) \\
r_f^* &\approx 0.05286 .
\end{aligned}
$$

The investor is thus indifferent between his optimal risky portfolio and a safe return of $5.29\%$

## Effects of changes to the risk aversion parameter $a$ and to the risk-free rate $r_f$

As shown in *Section 2.3*, the tangency portfolio is not defined by the risk aversion parameter. Therefore increasing $a$ to 8 has no effect on the stock-bond split that makes up the tangency portfolio.

However, it affects the position the investor takes along the CAL in order to maximise his utility. Following the optimal weight formula from *Section 2.4*, $w^*$ is inversely proportional to $a$, and since neither $\mathbb{E}[r_T]$ nor $\sigma_T$ depends on risk aversion, doubling $a$ exactly halves the weight placed in the tangency portfolio
$$
a = 2\cdot 4 = 8 \qquad \Longrightarrow \qquad w^{*} = \frac{1}{2}(1.2014) = 0.6007 .
$$

This is also known as the two fund seperation theorem.

---

In contrast, changing the risk-free rate to $r_f = 0.04$ affects the Sharpe maximisation, which defines the tangency portfolio. The weight in stocks $w_A$ rises from $36.1\%$ to $49.6\%$, moving the tangency portfolio to nearly an even split between stock and bond.

This is because, while in absolute terms both assets lose an equal $1\%$ of excess return, the bond loses more in relative terms. Its premium is halved from $2\%$ to $1\%$, whereas the equity premium falls only from $7\%$ to $6\%$. As the volatilities remain unchanged, the bond loses more of its return-to-risk ratio. Consequently the portfolio Sharpe maximisation favours the stock portion more.

## Dominated assets

An asset is dominated by another asset if it provides a lower expected return while having an equal or higher amount of risk or if it provides a lower or equal expected return while having an higher amount of risk.

This is the case for asset 3 and asset 2 where both inequalities hold strictly
$$
\mathbb{E}[r_3] = 9\% < 10\% = \mathbb{E}[r_2]  \qquad \text{and} \qquad \sigma_3 = 16\% > 12\% = \sigma_2 .
$$
As such, asset 2 dominates asset 3 by offering a higher expected return while carrying a lower amount of risk.
It should be noted, that in portfolio allocation this does not necessarily result in the exclusion of asset 3, as the covariance between the assets is also a critical measure.

## Assets on the efficient frontier, given a no-short sale constraint

Following the no-short sale constraint, the expected portfolio return for $N$ assets, which are indexed by $i$, is
$$
\mathbb{E}[r_P] = \sum_{i=1}^N w_i\,r_i \qquad \sum_{i=1}^N w_i = 1 \qquad w_i \ge 0 .
$$

When a no-shortsale constraint is imposed, the highest achievable portfolio return is the portfolio that is fully invested in the highest returning asset. 
$$
\mathbb{E}[r_{P, max}] = 1 \cdot r_{i, max} .
$$ 
From the table above, the highest returning asset is asset 6. 

The efficient frontier consists of all constructable portfolios that provide the highest level of return for a given level of risk.
Since no other asset achieves a higher return and asset 6 is the only one achieving $15\%$, there exists no other portfolio that offers at least the same return for a lower or equal amount of risk. Asset 6 must therefore lie on the efficient frontier regardless of its volatility of $20\%$.

- Each strategy starts with a single dollar invested at the beginning of the cohort
- The weights of the first four strategies are annually rebalanced to keep the same distributions and comparability
- The target date fund is rebalanced annually along a linear glide path, shifting the equity weight from 90% to 40%
- Inflation, taxes and transaction costs are disregarded

## Quantitative and qualitative analysis of the historical performances

The following code was generated by Claude Opus 5 to load the above referenced dataset, calculate the rolling 35-year cohorts for the five strategies, and visualise the distribution of annualized return and risk.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Setup
# Basic inputs for the retirement portfolio analysis

# Length of the savings horizon (a 30-year-old retiring at 65)
HORIZON = 35

# Rebalancing rule for the four fixed-weight portfolios
# True = reset to the target weight every year, False = buy and hold (weights drift with returns)
# Taxes and transaction costs are disregarded, which is defensible for a US retirement account
REBALANCE = True

# Start and end equity weight of the target date fund's glide path
GLIDE_FROM, GLIDE_TO = 0.90, 0.40

# Path to the Damodaran spreadsheet of annual returns since 1928
DATA_PATH = "Damodaran_hist_returns.xls"


# Load the data
# The table of annual returns starts in row 20 of the "Returns by year" sheet,
# so the 19 rows above it (title, links, premium calculator) are skipped
df = pd.read_excel(DATA_PATH, sheet_name="Returns by year", skiprows=19)

# Keep only the rows with a numeric year, which drops the summary rows at the bottom
df = df[pd.to_numeric(df["Year"], errors="coerce").notna()].copy()
df["Year"] = df["Year"].astype(int)

# Rename the three series we need and keep them in decimals (they are not percentages)
df = df.rename(columns={
    "Year": "year",
    "S&P 500 (includes dividends)": "stock",   # price appreciation plus dividends
    "US T. Bond (10-year)": "bond",            # coupon plus price appreciation
    "3-month T.Bill": "bill",                  # risk-free rate, used for the Sharpe ratio
})
df = df[["year", "stock", "bond", "bill"]].dropna().sort_values("year").reset_index(drop=True)
```

```python
# Define the strategies
# Each strategy is described by the equity weight it targets in every year of the horizon

# The four fixed-weight portfolios hold the same target weight throughout
def fixed_path(w_equity, h=HORIZON):
    return np.full(h, w_equity)

# The target date fund moves linearly from 90% to 40% equity over the 35 years
def glide_path(h=HORIZON, w0=GLIDE_FROM, w1=GLIDE_TO):
    return np.linspace(w0, w1, h)

# The glide path only exists if the fund is traded back to its target every year,
# so the target date fund is always rebalanced regardless of the REBALANCE switch
STRATEGIES = {
    "100% Equity":      dict(path=fixed_path(1.00), rebalance=REBALANCE),
    "80/20":            dict(path=fixed_path(0.80), rebalance=REBALANCE),
    "60/40":            dict(path=fixed_path(0.60), rebalance=REBALANCE),
    "40/60":            dict(path=fixed_path(0.40), rebalance=REBALANCE),
    "Target date fund": dict(path=glide_path(),     rebalance=True),
}
ORDER = list(STRATEGIES.keys())

# Simulate one cohort
# Track the equity and bond sleeves separately so that the weights are allowed to drift
def simulate(rs, rb, w_path, rebalance):
    h = len(rs)

    # Start with one dollar split according to the first target weight
    Ve, Vb = w_path[0], 1 - w_path[0]

    values  = np.empty(h + 1)
    values[0] = 1.0
    weights = np.empty(h)

    for t in range(h):
        V = Ve + Vb

        # Trade back to the target weight at the start of the year, if rebalancing
        if rebalance:
            Ve, Vb = w_path[t] * V, (1 - w_path[t]) * V

        # Record the weight the portfolio actually held during the year
        weights[t] = Ve / (Ve + Vb)

        # Let each sleeve grow at its own return
        Ve *= (1 + rs[t])
        Vb *= (1 + rb[t])
        values[t + 1] = Ve + Vb

    # Annual portfolio returns implied by the value path
    rets = values[1:] / values[:-1] - 1
    return values, rets, weights

# Largest peak-to-trough loss experienced along the way
def max_drawdown(values):
    peak = np.maximum.accumulate(values)
    return float((values / peak - 1).min())

# Run every cohort
# Using many cohorts avoids letting the market environment of one retirement date drive the result
def run_cohorts(df, strategies=STRATEGIES, h=HORIZON):
    rows = []
    starts = df["year"].values[:len(df) - h + 1]

    for s0 in starts:
        i  = df.index[df["year"] == s0][0]
        rs = df["stock"].values[i:i + h]
        rb = df["bond"].values[i:i + h]
        rf = df["bill"].values[i:i + h]

        for name, cfg in strategies.items():
            values, rets, weights = simulate(rs, rb, cfg["path"], cfg["rebalance"])
            tw = values[-1]
            rows.append(dict(
                cohort_start = s0,
                cohort_end   = s0 + h - 1,
                strategy     = name,
                terminal     = tw,                      # terminal wealth per dollar invested
                cagr         = tw ** (1 / h) - 1,       # annualised return over the cohort
                vol          = rets.std(ddof=1),        # volatility of the annual returns
                sharpe       = (rets - rf).mean() / rets.std(ddof=1),
                max_dd       = max_drawdown(values),
                final_w_eq   = weights[-1],             # equity weight reached by retirement
            ))
    return pd.DataFrame(rows)

# Summarise the distribution across cohorts.
# Terminal wealth is shown as a multiple of the dollar invested, the rest as percentages
def summarise(res, transpose=True):
    g = res.groupby("strategy")
    out = pd.DataFrame({
        "Median wealth":   g["terminal"].median().map("{:,.1f}x".format),
        "p10":             g["terminal"].quantile(0.10).map("{:,.1f}x".format),
        "p90":             g["terminal"].quantile(0.90).map("{:,.1f}x".format),
        "Worst":           g["terminal"].min().map("{:,.1f}x".format),
        "Best":            g["terminal"].max().map("{:,.1f}x".format),
        "Median return":   g["cagr"].median().map("{:.1%}".format),
        "Worst return":    g["cagr"].min().map("{:.1%}".format),
        "Volatility":      g["vol"].median().map("{:.1%}".format),
        "Sharpe":          g["sharpe"].median().map("{:.3f}".format),
        "Median drawdown": g["max_dd"].median().map("{:.1%}".format),
        "Worst drawdown":  g["max_dd"].min().map("{:.1%}".format),
    }).loc[ORDER]
    out.index.name = None
    return out.T if transpose else out

# Share of cohorts in which each strategy ended up ahead of the all-equity portfolio
def win_rate(res, benchmark="100% Equity"):
    p = res.pivot(index="cohort_start", columns="strategy", values="terminal")
    share = p[ORDER].gt(p[benchmark], axis=0).mean()
    return share.map("{:.1%}".format).rename(f"beat {benchmark}").to_frame()

# Plot the results
def plot_all(res):
    from matplotlib.ticker import NullFormatter

    p = res.pivot(index="cohort_start", columns="strategy", values="terminal")[ORDER]

    fig, ax = plt.subplots(1, 2, figsize=(15, 5.5))

    # Spread of terminal wealth across all cohorts
    ax[0].boxplot([p[c] for c in ORDER], tick_labels=ORDER, showmeans=True)
    ax[0].set_ylabel(f"Terminal wealth per $1 after {HORIZON} years (log)")
    ax[0].set_title("Distribution across cohorts")
    ax[0].tick_params(axis="x", rotation=20)
    ax[0].grid(True, alpha=0.3)

    # The same outcomes ordered by the year the cohort started saving
    for c in ORDER:
        ax[1].plot(p.index, p[c], label=c, lw=1.4)
    ax[1].set_xlabel("Cohort start year")
    ax[1].set_ylabel("Terminal wealth per $1 (log)")
    ax[1].set_title("Outcome by retirement cohort")
    ax[1].legend(fontsize=8)
    ax[1].grid(True, alpha=0.3)

    # A log scale keeps equal proportional differences equally spaced
    def log_ticks(lo, hi):
        base = np.array([1, 1.5, 2, 3, 4, 5, 7])
        cand = np.concatenate([base * 10.0 ** e for e in range(-2, 4)])
        return cand[(cand >= lo * 0.95) & (cand <= hi * 1.05)]

    ticks = log_ticks(p.values.min(), p.values.max())
    for a in (ax[0], ax[1]):
        a.set_yscale("log")
        if len(ticks) >= 3:
            a.set_yticks(ticks)
            a.set_yticklabels([f"{t:g}x" for t in ticks])
        a.yaxis.set_minor_formatter(NullFormatter())

    fig.tight_layout()

    # Realised risk against realised return, one point per cohort
    fig2, ax2 = plt.subplots(figsize=(9, 5))
    for c in ORDER:
        sub = res[res.strategy == c]
        ax2.scatter(sub["vol"] * 100, sub["cagr"] * 100, s=12, alpha=0.6, label=c)
    ax2.set_xlabel("Realised volatility over the cohort (%)")
    ax2.set_ylabel("Realised annualised return (%)")
    ax2.set_title("Risk and return, one point per cohort")
    ax2.legend(fontsize=8)
    ax2.grid(True, alpha=0.3)

    plt.show()
```

```python
# Run the simulation over every cohort and summarise the outcomes
res = run_cohorts(df)

print(f"Data: {df.year.min()}-{df.year.max()} ({len(df)} years), "
      f"{res.cohort_start.nunique()} overlapping {HORIZON}-year cohorts, "
      f"rebalancing: {REBALANCE}")

summarise(res)
```

Looking at the table, equities dominate on wealth multiples: a median of 39.0x, against 30.4x for the 80/20, 25.9x for the target date fund, 21.3x for the 60/40 and 14.9x for the 40/60 portfolio. The second table shows that the all-equity portfolio achieved the highest terminal wealth in 62 of the 64 cohorts, the two outlier cohorts were won by the target date fund. Even the worst all-equity cohort, 1929 to 1963, still returned 15.9x, which exceeds the median outcome of the 40/60 portfolio. 

At the same time equities showed the highest volatility of all strategies, at 17.2% against 8.9% for the 40/60, as well as the largest median drawdown of -36.5% and the largest single drawdown of -64.8%.

All in all, this results in the Sharpe ratio improving monotonically as the bond weight is increased, from 0.446 for the all-equity portfolio to 0.473 for the 40/60 portfolio. The diversified portfolios are therefore more efficient on a risk-adjusted basis.

## Portfolio recommendation for a 30-year-old and a 60-year-old investor

Given the analysis above the 30-year investor, with an investment horizon of 35-years, is recommended to follow an all equity portfolio. It delivered the highest terminal wealth in 62 of the 64 cohorts, or 96.9%, with its worst cohort even outperforming the median outcome of the 40/60 portfolio. In addition, it is the easiest to maintain for a retail investor, since holding a single index fund requires no rebalancing at all. Finally, the analysis above ignores transaction costs and taxable events, which would further favor the all equity portfolio. 
While a point could be made for holding the portfolio with the highest Sharpe ratio and leveraging it to the desired level of expected return. However, this strategy is less practical for a retail investors as it requires borrowing funds.

For the 60-year-old the horizon to retirement is only about five years, and the objective shifts to maintaining wealth and potentially financing continious withdrawals. Repeating the analysis over five-year cohorts shows the drastic influence of the investment horizon. The all-equity portfolio still produced the highest terminal wealth in 71.3% of cohorts, but it ended below its starting value in 11.7% of them, with a worst outcome of 0.5x. The 40/60 portfolio never fell below 0.9x. A loss of the size, as seen in the equity portfolio, shortly before retirement cannot be recovered and it would force withdrawals at depressed prices. The 40/60 portfolio is therefore recommended for the 60-year-old, on the basis of the lowest volatility and the smallest drawdowns of the five strategies.

## Potential caveats in the analysis of these portfolios

- Two adjacent cohorts share 34 of their 35 years, so the effective number of observations is far smaller than 64 and the percentiles do not carry the same weight as if drawn from an independent sample. The analysis also rests on a single realisation of history for one country. In fact, the U.S. equity market was among the most successful of the twentieth century, so resting the analysis solely on it does not represent the experience of a randomly chosen investor in 1928 (Goetzmann & Jorion, 1997).

- The analysis models a single lump sum invested at the start, whereas a real saver contributes gradually and later withdraws, which changes the exposure to the sequence in which returns arrive.

- Transaction costs, fund fees and the effort of rebalancing are ignored, all of which count against the portfolios that require annual trading. Taxes are ignored as well, which is defensible for a U.S. retirement account but not for a taxable one (Fidelity Investments, 2026). Finally, the glide path of the target date fund is assumed to be linear, whereas e.g. a stepped path would produce different results.

## Construction of a risk-matched comparison between the all equity portfolio and the 60/40 strategy

The following code cells were generated by Claude Opus 5.

```python
# Risk matching
# The diversified portfolio is levered until its volatility equals that of equities.
# Borrowing happens at the T-bill rate, so the levered portfolio returns
# r_f + L * (r_div - r_f). Because the T-bill rate barely varies, solving
# sigma(levered) = sigma(equity) reduces to the ratio of the two volatilities.
# W_EQ is defined once here and reused below, so the two cannot drift apart
W_EQ = 0.60

r_equity = df["stock"].to_numpy(float)
r_bill   = df["bill"].to_numpy(float)
r_div    = (W_EQ * df["stock"] + (1 - W_EQ) * df["bond"]).to_numpy(float)

L = r_equity.std(ddof=1) / r_div.std(ddof=1)
r_matched = r_bill + L * (r_div - r_bill)

# Dollar positions per $1 of investor capital
positions = pd.Series({
    "Stocks": W_EQ * L,
    "Bonds": (1 - W_EQ) * L,
    "T-bills (negative = borrowing)": 1 - L,
})

print(f"Leverage multiple L = {L:.4f}\n")
print(positions.map("{:.1%}".format).to_string())
print(f"Net exposure = {positions.sum():.1%}\n")
print(f"Equity volatility       = {r_equity.std(ddof=1):.2%}")
print(f"Risk-matched volatility = {r_matched.std(ddof=1):.2%}")

# The match is exact up to the small variation in the T-bill rate itself
assert np.isclose(r_equity.std(ddof=1), r_matched.std(ddof=1), rtol=0.01)
```

```python
# Apply the same leverage multiple in every cohort and compare against 100% equity
rows = []

for i in range(len(df) - HORIZON + 1):
    s = df.iloc[i:i + HORIZON]

    cohort_bill    = s["bill"].to_numpy(float)
    cohort_equity  = s["stock"].to_numpy(float)
    cohort_div     = (W_EQ * s["stock"] + (1 - W_EQ) * s["bond"]).to_numpy(float)
    cohort_matched = cohort_bill + L * (cohort_div - cohort_bill)

    # The leverage is estimated on the full sample, so check that the match still
    # roughly holds inside each cohort rather than assuming it does
    assert 0.8 < cohort_matched.std(ddof=1) / cohort_equity.std(ddof=1) < 1.25

    for name, r in [("100% Equity", cohort_equity),
                    ("Risk-matched 60/40", cohort_matched)]:
        values = np.r_[1.0, np.cumprod(1 + r)]
        excess = r - cohort_bill
        rows.append(dict(
            cohort_start = int(s["year"].iloc[0]),
            strategy     = name,
            terminal     = values[-1],
            cagr         = values[-1] ** (1 / HORIZON) - 1,
            vol          = r.std(ddof=1),
            sharpe       = excess.mean() / excess.std(ddof=1),
            max_dd       = max_drawdown(values),
        ))

rm_res = pd.DataFrame(rows)

g = rm_res.groupby("strategy")
rm_summary = pd.DataFrame({
    "Median wealth":   g["terminal"].median().map("{:,.1f}x".format),
    "Worst wealth":    g["terminal"].min().map("{:,.1f}x".format),
    "Median return":   g["cagr"].median().map("{:.2%}".format),
    "Worst return":    g["cagr"].min().map("{:.2%}".format),
    "Volatility":      g["vol"].median().map("{:.2%}".format),
    "Sharpe":          g["sharpe"].median().map("{:.3f}".format),
    "Median drawdown": g["max_dd"].median().map("{:.1%}".format),
    "Worst drawdown":  g["max_dd"].min().map("{:.1%}".format),
}).reindex(["100% Equity", "Risk-matched 60/40"])
rm_summary.index.name = None

rm_summary.T
```

```python
# How often the risk-matched portfolio delivered the higher annualised return,
# and how closely the full-sample leverage matched risk within individual cohorts
cagr_comparison = rm_res.pivot(index="cohort_start", columns="strategy", values="cagr")
vol_comparison  = rm_res.pivot(index="cohort_start", columns="strategy", values="vol")

win = (cagr_comparison["Risk-matched 60/40"] > cagr_comparison["100% Equity"]).mean()
ratio = vol_comparison["Risk-matched 60/40"] / vol_comparison["100% Equity"]

print(f"The risk-matched 60/40 has the higher annualised return in {win:.1%} "
      f"of the {len(cagr_comparison)} cohorts.")
print(f"Per-cohort volatility ratio: min {ratio.min():.2f}, "
      f"median {ratio.median():.2f}, max {ratio.max():.2f}")

ax = (100 * cagr_comparison).plot(figsize=(10, 5), linewidth=1.6)
ax.set_title("Risk-matched 60/40 versus 100% equity")
ax.set_xlabel("Cohort start year")
ax.set_ylabel(f"{HORIZON}-year annualised return (%)")
ax.grid(True, alpha=0.3)
plt.show()
```

The comparison in the previous sections also do not account for risk-adjusted returns. The diversified portfolios accumulate less wealth because they take less risk. In fact, they exhibit higher Sharpe ratios than the all equity portfolio. Following the AQR reading, the diversified portfolio should therefore be leveraged until its volatility matches that of the all-equity portfolio, to ensure a fair comparison.

We construct a risk adjusted portfolio which borrows at the T-bill rate and invests the proceeds in the 60/40 portfolio, so its return is $r_f + L(r_{60/40} - r_f)$. Here $L$ is the leverage multiple, which is needed to match the volatilities

$$
L = \frac{\sigma_{\text{equity}}}{\sigma_{60/40}} = \frac{19.40\%}{12.12\%} = 1.60 .
$$

Per dollar of capital, the risk-matched portfolio therefore holds 96.1% in stocks and 64.0% in bonds, financed by borrowing 60.1% at the T-bill rate.

Over the 64 cohorts the risk-matched portfolio delivers a median annualised return of 11.46% against 11.04% for the all-equity portfolio, and a median terminal wealth of 44.6x against 39.0x. It produced the higher annualised return in 62.5% of the cohorts, and improved the bottom cases, its weakest cohort returned 8.94% and 20.0x, against 8.22% and 15.9x for equities. This the result was implied by the higher Sharpe ratio, which were earlier discussed.

However, the risk-adjusted diversified portfolio does not win on every measure. The worst drawdown of the risk-matched portfolio is slightly smaller than that of equities, at -62.6% against -64.8%, but its median drawdown is larger, at -39.5% against -36.5%.

**Limitations:** The analysis assumes the investor can borrow at the T-bill rate, which the typical retail investor can not do. That borrowing spread is subtracted directly from the 0.42 percentage point advantage shown above, which might be enough to completely erase it. Leverage also introduces risks additional risks, such as margin calls. The strategy requires annual rebalancing of a levered position, so transaction costs and taxes would likely weigh more heavily on it compared to a simple equity index fund holding.

The leverage multiple is estimated once over the full sample and then applied to every cohort, so volatility is matched exactly over 1928 to 2025 but only approximately within individual cohorts. Finally an investor could also use a different risk measure, such as the largest loss rather than the standard deviation and on that measure the levered portfolio is the riskier choice.
