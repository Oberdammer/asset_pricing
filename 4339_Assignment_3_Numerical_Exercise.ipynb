## 1. Data and monthly excess returns

**Fund returns.** For each fund we use the monthly *adjusted* closing price from Yahoo Finance. The adjusted price assumes that all dividends and capital gains distributions are reinvested, so the simple monthly return
$$
r_{i,t} = \frac{P^{adj}_{i,t}}{P^{adj}_{i,t-1}} - 1
$$
is the fund's total return. Because a fund's net asset value is reported after the deduction of the management fee and other expenses, these are **net-of-fee returns**, i.e. the returns that an investor in the fund actually earned. The price series starts in December 2001 so that the first return is the one for January 2002. Yahoo labels each monthly bar with the first day of the month (e.g. 2026-07-01), while the price itself is the closing price on the last trading day of that month, so the bar dated 2026-07-01 delivers the July 2026 return.

**Factors and risk-free rate.** From the Kenneth French data library we use the monthly file *Fama/French 3 Factors* and the *Momentum Factor*:

- $r_{f,t}$ (RF): one-month Treasury bill return,
- $r_{M,t}-r_{f,t}$ (Mkt-RF): excess return of the value-weighted market portfolio of all CRSP firms listed on the NYSE, AMEX and NASDAQ,
- $SMB_t$ (Small Minus Big): return on small-cap stocks minus large-cap stocks (size factor),
- $HML_t$ (High Minus Low): return on high book-to-market (value) stocks minus low book-to-market (growth) stocks (value factor),
- $Mom_t$: return on stocks with high prior 2-12 month returns minus stocks with low prior returns (momentum factor of Carhart, 1997).

The file reports returns in percent, so all values are divided by 100. SMB, HML and Mom are returns on zero-investment long-short portfolios and are therefore already "excess" returns.

**Excess returns.** After matching both data sets by calendar month, the monthly excess return of fund $i$ is
$$
r^e_{i,t} = r_{i,t} - r_{f,t}.
$$

The code below downloads the fund prices from Yahoo Finance and the factors from the Kenneth French data library and saves them as `fund_prices_yahoo.csv` and `ffc4_factors_french.csv` next to the notebook. If the two files already exist, they are loaded instead of downloaded again. The results below are based on data downloaded on 13 October 2026.

```python
DATA_DIR = Path.cwd()
PRICE_FILE = DATA_DIR / "fund_prices_yahoo.csv"
FACTOR_FILE = DATA_DIR / "ffc4_factors_french.csv"
FRENCH_URL = "https://mba.tuck.dartmouth.edu/pages/faculty/ken.french/ftp/"

PRICE_START, PRICE_END = "2001-12-01", "2026-08-01"            # Yahoo end date is exclusive
SAMPLE_START, SAMPLE_END = pd.Period("2002-01", "M"), pd.Period("2026-07", "M")


def download_prices(tickers, start, end):
    """Monthly adjusted closing prices from Yahoo Finance."""
    install_and_import("yfinance", "yf")
    raw = yf.download(tickers, start=start, end=end, interval="1mo",
                      auto_adjust=False, actions=False, progress=False)
    px = raw["Adj Close"][tickers].dropna(how="all")
    px.index = pd.to_datetime(px.index).to_period("M")          # 2026-07-01 -> 2026-07
    px.index.name = "date"
    return px


def download_french(zip_name):
    """Monthly table (in percent) from a zipped CSV file of the Kenneth French data library."""
    import io, zipfile, requests
    resp = requests.get(FRENCH_URL + zip_name, timeout=60)
    resp.raise_for_status()
    with zipfile.ZipFile(io.BytesIO(resp.content)) as z:
        text = z.read(z.namelist()[0]).decode("latin-1")
    header, rows = None, []
    for line in text.splitlines():
        parts = [s.strip() for s in line.split(",")]
        if header is None:                                       # first line like ",Mkt-RF,SMB,HML,RF"
            if len(parts) > 1 and parts[0] == "" and parts[1] != "":
                header = parts[1:]
            continue
        if len(parts) > 1 and parts[0].isdigit() and len(parts[0]) == 6:
            rows.append(parts)                                   # monthly row, e.g. "200201,-1.44,..."
        elif rows:                                               # end of the monthly block (annual data follow)
            break
    df = pd.DataFrame([r[1:] for r in rows], index=[r[0] for r in rows], columns=header).astype(float)
    df.index.name = "date"
    return df


# ---- fund prices ----
if PRICE_FILE.exists():
    prices = pd.read_csv(PRICE_FILE, index_col="date")
    prices.index = pd.PeriodIndex(prices.index, freq="M")
    print(f"Fund prices loaded from local copy: {PRICE_FILE.name}")
else:
    prices = download_prices(FUNDS, PRICE_START, PRICE_END)
    prices.to_csv(PRICE_FILE, index_label="date")
    print("Fund prices downloaded from Yahoo Finance and saved.")

# ---- factors (in percent, as in the French library) ----
if FACTOR_FILE.exists():
    factors_pct = pd.read_csv(FACTOR_FILE, index_col="date")
    print(f"Factors loaded from local copy: {FACTOR_FILE.name}")
else:
    ff3 = download_french("F-F_Research_Data_Factors_CSV.zip")
    mom = download_french("F-F_Momentum_Factor_CSV.zip")
    factors_pct = ff3.join(mom, how="inner")
    factors_pct.to_csv(FACTOR_FILE, index_label="date")
    print("Factors downloaded from the Kenneth French data library and saved.")

factors_pct.index = pd.to_datetime(factors_pct.index.astype(str), format="%Y%m").to_period("M")
factors_pct.columns = [c.strip() for c in factors_pct.columns]

prices = prices.loc[SAMPLE_START - 1:SAMPLE_END, FUNDS]          # December 2001 to July 2026
factors = factors_pct.loc[SAMPLE_START:SAMPLE_END, ["Mkt-RF", "SMB", "HML", "Mom", "RF"]] / 100
prices.tail(3)
```

```python
rets = prices.pct_change().dropna()                              # simple monthly total returns
assert rets.index.equals(factors.index), "Fund and factor months do not line up"
assert rets.notna().all().all() and factors.notna().all().all(), "Missing observations"

rf = factors["RF"]
excess = rets.sub(rf, axis=0)                                    # fund excess returns
mkt = factors["Mkt-RF"]                                          # market excess return

print(f"Sample: {excess.index[0].strftime('%b %Y')} to {excess.index[-1].strftime('%b %Y')}, T = {len(excess)} months "
      f"(identical for all four funds and the factors)")
excess.head()
```

The market's total return is $(r_{M,t}-r_{f,t}) + r_{f,t}$.

- **Lower return and higher volatility: GLCGX.** GLCGX earned 0.78 pp p.a. less than the market (8.53%) while being more volatile (16.35%). In terms of mean and standard deviation, the market dominated this fund over the sample.
- **Lower return and lower volatility: STSEX.** STSEX earned 1.70 pp p.a. less than the market (7.61%), but was also clearly less volatile (13.27%) and had the mildest worst month (-12.3%, against -17.2% for the market). ## 2. Sharpe ratio and information ratio

**Sharpe ratio.** The Sharpe ratio measures the expected excess return per unit of total risk,
$$
\hat S_i = \frac{\bar r^e_i}{\hat\sigma_i},
$$
where $\bar r^e_i$ and $\hat\sigma_i$ are the sample mean and standard deviation of the monthly excess returns. It is the slope of the CAL obtained by combining the fund with T-bills, so it is the relevant measure if the fund is an investor's *entire* risky portfolio. With i.i.d. returns the mean scales with 12 and the standard deviation with $\sqrt{12}$, so $\hat S^{ann}_i = \sqrt{12}\,\hat S_i$.

**Information ratio relative to the market.** The information ratio measures performance relative to a benchmark, here the market. The active return is the difference between the fund return and the benchmark return, in which the risk-free rate cancels out,
$$
a_{i,t} = r_{i,t} - r_{M,t} = r^e_{i,t} - r^e_{M,t},
\qquad
\widehat{IR}_i = \frac{\bar a_i}{sd(a_i)},
$$
where the denominator $sd(a_i)$ is the **tracking error**. The IR is the average outperformance of the market per unit of risk taken by deviating from it, and is annualised with $\sqrt{12}$ as well. Since $sd(a_i)/\sqrt{T}$ is the standard error of the mean active return, the t-statistic for $H_0: \mathbb{E}[a_{i}] = 0$ (no out- or underperformance of the market) is simply the monthly information ratio multiplied by $\sqrt{T}$,
$$
t_{\bar a_i} = \frac{\bar a_i}{sd(a_i)/\sqrt{T}} = \widehat{IR}_i\,\sqrt{T} .
$$

**Appraisal ratio (CAPM).** The IR implicitly assumes that the fund has a market beta of one. As a complement, the regression-based appraisal ratio first removes the fund's market exposure,
$$
r^e_{i,t} = \alpha_i + \beta_i\, r^e_{M,t} + \varepsilon_{i,t},
\qquad
\widehat{AR}_i = \frac{\hat\alpha_i}{sd(\hat\varepsilon_i)} .
$$
Averaging the regression over the sample (OLS residuals have mean zero) links the two measures exactly:
$$
\bar a_i = \bar r^e_i - \bar r^e_M = \hat\alpha_i + (\hat\beta_i - 1)\,\bar r^e_M .
$$
The average active return thus consists of CAPM alpha plus a part that only reflects a market exposure above or below one.

```python
active = excess.sub(mkt, axis=0)                                 # a_t = r_i - r_M
t_active = active.mean() / (active.std() / np.sqrt(len(active)))   # = monthly IR x sqrt(T)
p_active = pd.Series(2 * stats.t.sf(t_active.abs(), df=len(active) - 1), index=FUNDS)
capm = {k: sm.OLS(excess[k], sm.add_constant(mkt)).fit() for k in FUNDS}

perf = pd.DataFrame({
    "Mean excess return (ann.)": excess.mean() * 12,
    "SD (ann.)": excess.std() * np.sqrt(12),
    "Sharpe ratio (monthly)": excess.mean() / excess.std(),
    "Sharpe ratio (ann.)": excess.mean() / excess.std() * np.sqrt(12),
    "Mean active return (ann.)": active.mean() * 12,
    "Tracking error (ann.)": active.std() * np.sqrt(12),
    "Information ratio (monthly)": active.mean() / active.std(),
    "Information ratio (ann.)": active.mean() / active.std() * np.sqrt(12),
    "t-stat mean active return": t_active,
    "p-value mean active return": p_active,
    "CAPM beta": pd.Series({k: capm[k].params["Mkt-RF"] for k in FUNDS}),
    "CAPM alpha (ann.)": pd.Series({k: capm[k].params["const"] * 12 for k in FUNDS}),
    "Beta effect, (beta - 1) x mean mkt excess (ann.)": pd.Series({k: (capm[k].params["Mkt-RF"] - 1) * mkt.mean() * 12 for k in FUNDS}),
    "Appraisal ratio (ann.)": pd.Series({k: capm[k].params["const"] / np.sqrt(capm[k].mse_resid) * np.sqrt(12) for k in FUNDS}),
})
perf.loc["MKT"] = [mkt.mean() * 12, mkt.std() * np.sqrt(12), mkt.mean() / mkt.std(), mkt.mean() / mkt.std() * np.sqrt(12),
                   0, 0, np.nan, np.nan, np.nan, np.nan, 1, 0, 0, np.nan]

pct_cols = ["Mean excess return (ann.)", "SD (ann.)", "Mean active return (ann.)", "Tracking error (ann.)",
            "CAPM alpha (ann.)", "Beta effect, (beta - 1) x mean mkt excess (ann.)"]
perf.rename(index=NAMES).T.style.format(
    lambda v: "" if pd.isna(v) else f"{v:.3f}"
).format(lambda v: f"{v:.2%}", subset=pd.IndexSlice[pct_cols, :])
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4.2))
xx = np.arange(len(FUNDS))

ax = axes[0]
sr = perf.loc[FUNDS, "Sharpe ratio (ann.)"]
ax.bar(xx, sr, color=[COL[k] for k in FUNDS], width=0.6)
ax.axhline(perf.loc["MKT", "Sharpe ratio (ann.)"], color=COL["MKT"], ls="--", lw=1.4,
           label=f"Market: {perf.loc['MKT', 'Sharpe ratio (ann.)']:.3f}")
for i, v in enumerate(sr):
    ax.text(i, v + 0.01, f"{v:.3f}", ha="center", fontsize=9)
ax.set_xticks(xx, FUNDS); ax.set_ylabel("Annualised Sharpe ratio"); ax.set_ylim(0, 0.72)
ax.set_title("Sharpe ratio", loc="left"); ax.legend(frameon=False, fontsize=8.5, loc="lower right")

ax = axes[1]
ir = perf.loc[FUNDS, "Information ratio (ann.)"]
ar = perf.loc[FUNDS, "Appraisal ratio (ann.)"]
ax.bar(xx, ir, color=[COL[k] for k in FUNDS], width=0.6)
ax.scatter(xx, ar, marker="D", s=46, color="#0b0b0b", zorder=3, label="Appraisal ratio (CAPM alpha)")
ax.axhline(0, color="grey", lw=0.8)
for i, v in enumerate(ir):
    ax.text(i + 0.33, v, f"{v:+.3f}", ha="left", va="center", fontsize=8.5)
ax.set_xticks(xx, FUNDS); ax.set_ylabel("Annualised ratio")
ax.set_title("Information ratio and appraisal ratio", loc="left")
from matplotlib.patches import Patch
from matplotlib.lines import Line2D
ax.legend(handles=[Patch(color="#b8b7b1", label="Information ratio vs. market (bars, fund colours)"),
                   Line2D([], [], marker="D", ls="", color="#0b0b0b", label="Appraisal ratio (CAPM alpha)")],
          frameon=False, fontsize=8.5, loc="lower left")
plt.tight_layout(); plt.show()
```


### Discussion

**Sharpe ratios.** None of the four funds achieved a higher Sharpe ratio than the market (0.610). TRBCX (0.575) and STSEX (0.574) come closest, followed by GLCGX (0.522) and DFSVX (0.499). An investor who held the market portfolio and T-bills, choosing the mix that gives the desired level of risk on the market's CAL, would therefore have obtained a better risk-return trade-off than with any one of these funds as the entire risky portfolio. It is interesting that TRBCX and STSEX reach almost the same Sharpe ratio in very different ways: TRBCX earns a high average excess return (9.85%) with high volatility (17.1%), whereas STSEX earns less (7.61%) with much lower volatility (13.3%). DFSVX has the highest average excess return of all (10.71%), but its volatility of 21.5% is so high that it ends up with the lowest Sharpe ratio.

**Information ratios.** Relative to the market, TRBCX (IR = 0.089) and DFSVX (IR = 0.124) outperformed on average by 0.54% and 1.40% p.a., with tracking errors of 6.1% and 11.3%. GLCGX (IR = -0.173) and STSEX (IR = -0.288) underperformed by 0.78% and 1.70% p.a. All information ratios are small. The t-statistics of the mean active returns reported in the table range from -1.43 (STSEX, $p = 0.15$) to 0.61 (DFSVX), so none of the funds out- or underperformed the market significantly.

**Why the IR alone is not evidence of skill.** The decomposition $\bar a_i = \hat\alpha_i + (\hat\beta_i - 1)\bar r^e_M$ shows that most of the active returns are explained by market beta and not by alpha. DFSVX has a beta of 1.22, so its higher market exposure alone accounts for +2.00% p.a. of active return, which more than explains the +1.40% it actually earned; its CAPM alpha is -0.60%. Conversely, STSEX's underperformance of 1.70% p.a. comes entirely from its low beta of 0.80 (-1.83% p.a.), while its CAPM alpha is slightly positive (+0.14%). In a period in which the market earned a high premium (9.31% p.a.), any fund with $\beta > 1$ looks good and any fund with $\beta < 1$ looks bad against the market, even though an investor could create the same market exposure with leverage or T-bills at no cost. The appraisal ratios, which correct for beta, are all close to zero (between -0.23 for GLCGX and +0.03 for STSEX). Whether the remaining small alphas reflect skill or only exposure to other systematic factors such as size and value is what we test in the next section.

## 3. Evidence of fund manager skill: factor regressions

**Background.** A manager has skill if the fund earns a higher return than an investor could obtain with a passive portfolio that has the same systematic risk exposures. In a factor model, this abnormal return is measured by the intercept $\alpha_i$ of a time-series regression of the fund's excess return on the factor returns. Since the right-hand side variables are returns on traded (excess-return) portfolios, $\sum_k \beta_{i,k} f_{k,t}$ is the return of a passive replicating portfolio of the factors and T-bills, and $\alpha_i$ is the average return of the fund in excess of that replicating portfolio. If the factor model correctly describes expected returns, $\alpha_i = 0$ for a fund without skill.

**Fama-French three-factor model (FF3):**
$$
r^e_{i,t} = \alpha_i + \beta_{i,M}\,(r_{M,t}-r_{f,t}) + \beta_{i,S}\,SMB_t + \beta_{i,H}\,HML_t + \varepsilon_{i,t}
$$

**Fama-French-Carhart four-factor model (FFC4):**
$$
r^e_{i,t} = \alpha_i + \beta_{i,M}\,(r_{M,t}-r_{f,t}) + \beta_{i,S}\,SMB_t + \beta_{i,H}\,HML_t + \beta_{i,Mom}\,Mom_t + \varepsilon_{i,t}
$$

The factor loadings describe the fund's investment style: $\beta_{i,S} > 0$ indicates a tilt towards small stocks, $\beta_{i,H} > 0$ a tilt towards value stocks and $\beta_{i,H} < 0$ towards growth stocks, and $\beta_{i,Mom} > 0$ a tilt towards past winners. A manager who simply follows a size, value or momentum style earns the corresponding factor premia, but this is not skill in the sense of the models because the same exposure can be obtained passively.

**Hypothesis test.** A positive estimated alpha alone is not enough, since it could be the result of luck. For each fund we test
$$
H_0: \alpha_i = 0 \quad \text{against} \quad H_1: \alpha_i \neq 0,
\qquad
t_{\alpha_i} = \frac{\hat\alpha_i}{SE(\hat\alpha_i)},
$$
and reject $H_0$ at the conventional two-sided 5% level if $|t| > 1.97$ ($p < 0.05$). Evidence of skill requires a significantly *positive* alpha. Our main results are based on the standard OLS standard errors. Because monthly fund returns may show heteroskedasticity and some autocorrelation, we additionally report the p-value based on Newey-West (HAC) standard errors with 6 lags as a robustness check. Alphas are monthly in the regressions and annualised by multiplying by 12. To see how alpha changes as factors are added, we also estimate the CAPM regression from Section 2 as a baseline.

```python
MODELS = {"CAPM": ["Mkt-RF"], "FF3": ["Mkt-RF", "SMB", "HML"], "FFC4": ["Mkt-RF", "SMB", "HML", "Mom"]}

res, res_hac = {}, {}
for m, cols in MODELS.items():
    X = sm.add_constant(factors[cols])
    for k in FUNDS:
        res[m, k] = sm.OLS(excess[k], X).fit()
        res_hac[m, k] = sm.OLS(excess[k], X).fit(cov_type="HAC", cov_kwds={"maxlags": 6})


def alpha_table(model):
    rows = {}
    for k in FUNDS:
        r, h = res[model, k], res_hac[model, k]
        lo, hi = r.conf_int().loc["const"] * 12
        rows[k] = {
            "alpha (monthly)": r.params["const"],
            "alpha (ann.)": r.params["const"] * 12,
            "SE alpha (ann.)": r.bse["const"] * 12,
            "95% CI alpha (ann.)": f"[{lo:.2%}, {hi:.2%}]",
            "t(alpha)": r.tvalues["const"],
            "p-value": r.pvalues["const"],
            "p-value (Newey-West)": h.pvalues["const"],
            **{f"beta {c}": r.params[c] for c in MODELS[model]},
            **{f"t({c})": r.tvalues[c] for c in MODELS[model]},
            "R-squared": r.rsquared,
            "Adj. R-squared": r.rsquared_adj,
            "Residual SD (ann.)": np.sqrt(r.mse_resid * 12),
        }
    return pd.DataFrame(rows)


def show(tab):
    pct = ["alpha (ann.)", "SE alpha (ann.)", "Residual SD (ann.)"]
    return tab.rename(columns=NAMES).style.format(
        lambda v: v if isinstance(v, str) else f"{v:.4f}"
    ).format(lambda v: f"{v:.2%}", subset=pd.IndexSlice[pct, :]
    ).format(lambda v: f"{v:.3%}", subset=pd.IndexSlice[["alpha (monthly)"], :])


def regression_output(model):
    out = summary_col([res[model, k] for k in FUNDS], model_names=FUNDS, stars=True,
                      float_format="%.5f", info_dict={"N": lambda r: f"{int(r.nobs)}"},
                      regressor_order=["const"] + MODELS[model])
    print(f"{model} regressions of monthly fund excess returns (standard errors in parentheses)")
    print(out)
```

### (a) FF3 regressions

```python
regression_output("FF3")
```

```python
show(alpha_table("FF3"))
```

**Factor loadings.** The FF3 loadings match the funds' stated investment styles closely, and all are highly significant:

- **TRBCX** ($\beta_M = 1.09$, $\beta_S = -0.08$, $\beta_H = -0.40$) and **GLCGX** ($\beta_M = 1.06$, $\beta_S = -0.07$, $\beta_H = -0.27$) are large-cap growth funds: slightly more market risk than the market, a small tilt towards large firms and a strong negative loading on value, i.e. a tilt towards growth stocks.
- **STSEX** ($\beta_M = 0.85$, $\beta_S = -0.29$, $\beta_H = +0.10$) holds very large, somewhat value-oriented firms, with a below-average market beta.
- **DFSVX** ($\beta_M = 1.03$, $\beta_S = 0.85$, $\beta_H = 0.64$) has by far the largest loadings on both SMB and HML, as expected for a small-cap value fund.

Adding SMB and HML increases the explanatory power considerably compared with the CAPM, especially for DFSVX, whose $R^2$ rises from 0.75 to 0.97. With $R^2$ between 0.89 and 0.97, the three factors capture almost all of the funds' return variation, and the remaining residual (non-factor) volatility is only 3.3% to 4.4% p.a.

**Alphas.** No fund has a significant FF3 alpha. The annualised estimates are +0.04% for TRBCX ($t = 0.05$), -1.08% for GLCGX ($t = -1.57$, $p = 0.12$), -0.13% for STSEX ($t = -0.14$) and +0.07% for DFSVX ($t = 0.09$). We cannot reject $H_0: \alpha_i = 0$ at the 5% level (nor at the 10% level) for any fund, and the Newey-West p-values lead to the same conclusion. The only alpha estimate of economically meaningful size, GLCGX's, is *negative*.

Compared with the CAPM, the FF3 alpha of DFSVX increases from -0.60% to +0.07% p.a. Under the CAPM, part of the fund's small-cap value exposure is picked up by a high market beta of 1.22. Once size and value are controlled for directly, the market beta falls to 1.03 and DFSVX's performance is fully explained by its factor exposures. TRBCX's alpha is close to zero under both models.

### (b) FFC4 regressions

```python
regression_output("FFC4")
```

```python
show(alpha_table("FFC4"))
```

**Momentum loadings.** The loadings on the momentum factor are small for all four funds: +0.014 for TRBCX ($t = 0.8$), +0.035 for GLCGX ($t = 2.6$), -0.044 for STSEX ($t = -2.4$) and -0.014 for DFSVX ($t = -1.0$). Two of these loadings are statistically significant at the 5% level. GLCGX has a significant positive loading, which may come from a more systematic, algorithmic stock selection process, as the "Insights" in the fund's name suggests, if the models behind it tend to favour stocks with strong recent price trends. STSEX has a significant negative loading, which is consistent with a low-turnover portfolio that does not chase recent winners. For TRBCX and DFSVX, the momentum loadings are insignificant. Even where they are significant, the loadings are economically small: with a momentum premium of 2.03% p.a. over the sample, GLCGX's loading adds only about 0.07% p.a. to its expected return and STSEX's loading subtracts about 0.09% p.a. The other loadings and the $R^2$ are virtually unchanged from FF3 for all four funds, so none of them follows a momentum strategy in any meaningful sense.

**Alphas.** Adding momentum therefore changes the alphas only slightly: TRBCX -0.04% ($t = -0.05$), GLCGX -1.28% ($t = -1.86$, $p = 0.063$), STSEX +0.12% ($t = 0.13$) and DFSVX +0.14% ($t = 0.20$) p.a. Since the momentum factor earned a premium of 2.03% p.a. over the sample, GLCGX's positive momentum exposure explains part of its return, so its alpha becomes more negative. Again, no fund has an alpha that is significantly different from zero at the 5% level. GLCGX's negative alpha is significant at the 10% level (Newey-West $p = 0.066$), which is weak evidence of *underperformance* and not of skill.

The chart below compares the annualised alpha estimates and their 95% confidence intervals across the three models.

**Robustness: subperiods.** A full-sample alpha of zero could hide skill in one part of the sample that disappears later, for example because a successful fund attracts inflows and grows too large (Berk and Green, 2004). We therefore re-estimate the FFC4 regression separately for the first and second half of the sample.

```python
halves = {"2002-01 to 2013-12": (pd.Period("2002-01", "M"), pd.Period("2013-12", "M")),
          "2014-01 to 2026-07": (pd.Period("2014-01", "M"), pd.Period("2026-07", "M"))}
sub = {}
for lab, (s, e) in halves.items():
    X = sm.add_constant(factors.loc[s:e, MODELS["FFC4"]])
    for k in FUNDS:
        r = sm.OLS(excess.loc[s:e, k], X).fit()
        sub[k] = sub.get(k, {}) | {(lab, "alpha (ann.)"): r.params["const"] * 12,
                                   (lab, "t(alpha)"): r.tvalues["const"],
                                   (lab, "p-value"): r.pvalues["const"]}
sub = pd.DataFrame(sub).T
sub.columns = pd.MultiIndex.from_tuples(sub.columns, names=["FFC4, subperiod", ""])
sub.rename(index=NAMES).style.format("{:.3f}").format(
    "{:.2%}", subset=[c for c in sub.columns if c[1] == "alpha (ann.)"])
```

The subperiod results confirm the full-sample picture. TRBCX, STSEX and DFSVX have insignificant alphas with changing signs in both halves. GLCGX's underperformance is concentrated in the first half, where its FFC4 alpha of -2.13% p.a. is significant ($t = -2.65$); in the second half it is -0.61% p.a. and insignificant. There is no subperiod in which any fund delivers a significantly positive alpha.



### (c) Conclusion: is there evidence of fund manager skill?

**No.** Neither the FF3 nor the FFC4 model provides evidence of manager skill for any of the four funds. All estimated alphas are statistically indistinguishable from zero at the 5% level, three of them are within $\pm 0.15\%$ p.a. of zero under both models, and the only economically sizeable alpha (GLCGX, about -1.1% to -1.3% p.a.) is negative and marginally significant at the 10% level under FFC4. The results are the same with Newey-West standard errors and in both halves of the sample, apart from GLCGX's significant underperformance before 2014.

The funds' performance is instead almost entirely explained by their exposures to the market, size, value and momentum factors. This is consistent with the results from Section 2. The positive information ratios of TRBCX and DFSVX are the result of market betas above one and, for DFSVX, of its small-cap value style, not of superior stock selection. An investor could have replicated the risk and return of each fund with a passive portfolio of the factors, which is precisely the case when $\alpha = 0$.

There are several points to consider when interpreting these results:

- **Net versus gross alpha.** The fund returns are net of fees. An alpha of zero after fees implies a small positive alpha *before* fees, roughly equal to the annual expense ratio of the fund (a fraction of a percent per year for these funds). The managers may therefore have some ability to generate returns, but it is just enough to cover their fees, so that investors do not benefit from it. This is consistent with the "arithmetic of active management" (Sharpe, 1991): since passive investors hold the market portfolio, active investors in aggregate must hold it as well, so the average active fund earns a gross alpha of about zero and a negative net alpha after fees. It is also consistent with Berk and Green (2004), where fund flows compete away any net alpha.
- **Statistical power.** The standard errors of the annualised alphas are between 0.7% and 0.9%, so the 95% confidence intervals are about $\pm 1.3\%$ to $\pm 1.8\%$ p.a. wide. Even with almost 25 years of data, a true alpha of around +1% p.a., which would be economically very valuable, could not be distinguished from zero. The absence of evidence of skill is therefore not the same as evidence of the absence of skill, although the point estimates are centred close to zero.
- **The factor benchmark is not free.** SMB, HML and Mom are returns on long-short paper portfolios that ignore trading costs, short-selling costs and fees. For DFSVX, an alpha of zero relative to these factors means that the fund delivered its size and value exposure without losing return to costs, which still has value for investors who want this exposure. Measured against investable index funds instead of paper portfolios, its performance would probably look better.
- **Selection and multiple testing.** The four funds were chosen ex post among funds that survived the full 25-year period, which biases the sample towards successful funds. Moreover, with four funds and two models, finding one alpha that is significant at the 10% level is not unusual even if all true alphas were zero. Both effects make it easier, not harder, to find positive alphas by chance, which strengthens the conclusion that there is no evidence of skill.

In sum, the regressions suggest that what investors pay for in these funds is not stock-picking skill but systematic factor exposure: market risk for all four funds, a growth tilt for TRBCX and GLCGX, a large-cap value tilt for STSEX and a small-cap value tilt for DFSVX.
