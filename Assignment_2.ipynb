# The CAPM

Consider an economy with a portfolio of risky assets and a risk-free asset in which the CAPM holds. The following table gives the expected return $\mathbb{E}[r]$ and the standard deviation $\sigma$ of these two assets.

Portfolio & Expected Return & Standard Deviation \\
\midrule
Risk free ($r_f$) & 2\% & 0\% \\
Market ($r_M$) & 8\% & 20\% \\

## Computation of Beta $\beta$ for an asset that correlates with the market.

Assume asset $A$ has standard deviation $\sigma_A = 10\%$ and correlation $\rho_{A,M}=0.5$ with the market.
Beta measures the sensitivity of asset $A$'s returns relative to the market $M$. It is defined as
$$
\beta_A = \frac{\sigma_{A,M}}{\sigma_M^2} \qquad \text{where} \qquad \sigma_{A,M} = \rho_{A,M} \sigma_A \sigma_M
$$

Substituting the values yields
$$
\beta_A = \frac{0.5 \cdot 0.1 \cdot 0.2}{0.2^2} = 0.25 .
$$
Asset A carries one quarter of the market's systematic risk.

---

## Deriving the risk-aversion coefficient under the CAPM

Under the CAPM all investors hold a combination of the risk-free asset and the tangency portfolio $T$, which is equal to the market portfolio. Therefore, $\mathbb{E}[r_T] = \mathbb{E}[r_M] = 8\%$ and $\sigma_T = \sigma_M = 20\%$.

The investor puts weight $w_M$ into the market portfolio and weight $1-w_M$ into the risk-free asset, with the goal to maximize his utility function
\begin{equation}
U = r_f + w_M (\mathbb{E}[r_M] - r_f)-\frac{a}{2}(w_M)^2\sigma_M^2 . \label{eq:CAPM}
\end{equation}

Since the risk-free asset has zero variance, the standard deviation of the investor's portfolio is
$$
\begin{aligned}
\sigma_P    &= w_M \sigma_M \\
w_M         &= \frac{\sigma_P}{\sigma_M} .
\end{aligned}
$$
Given that the portfolio standard deviation is $\sigma_P=0.1$, this results in
$$
w_M = \frac{0.10}{0.20} = 0.5 \qquad \text{and} \qquad w_f = 1-w_M = 0.5 .
$$

To find the maximum of the utility formula we differentiate \eqref{eq:CAPM} with respect to $w_M$ and set it equal to zero:
$$
\begin{aligned}
\frac{dU}{dw_M}   &= (\mathbb{E}[r_M]-r_f) - a\,w_M\,\sigma_M^2 = 0 \\
a               &= \frac{(\mathbb{E}[r_M]-r_f)}{w_M\,\sigma_M^2}
\end{aligned}
$$

Substituting the values yields
$$
a = \frac{0.08 - 0.02}{0.5 \cdot 0.2^2} = \frac{0.06}{0.02} = 3 .
$$
The investor's risk-aversion coefficient is 3.

# 2. CAPM with public and private equity

There are two broad classes of risky assets: private equity, $P$, and public equity, $U$. Their joint covariance matrix has the block structure

$$
\Sigma =
\begin{pmatrix}
\Sigma_P & \Sigma_{PU} \\
\Sigma_{PU}^{\mathsf T} & \Sigma_U
\end{pmatrix}.
$$

There are two investors: a private-equity (PE) fund and a public-equity mutual fund. Unless a subexercise imposes an additional restriction, we maintain the standard CAPM assumptions:

- Investors are price takers and choose portfolios using one-period mean--variance preferences.
- Investors have homogeneous beliefs about expected returns, variances, and covariances.
- A risk-free asset exists, and investors can borrow or lend at the common rate $r_f$.
- Assets are divisible, and there are no taxes, transaction costs, or other trading frictions apart from the restrictions stated in each subexercise.
- The supplies of the risky assets are fixed, so equilibrium market clearing requires investors collectively to hold all outstanding public and private equity.

## (a) Frictionless benchmark

Both the PE fund and the public-equity mutual fund can freely invest in public and private equity. Therefore, they face the same investment opportunity set. Under the standard CAPM assumptions, both investors hold combinations of the risk-free asset and the same tangency portfolio of risky assets.

In equilibrium, market clearing implies that this common tangency portfolio must be the value-weighted portfolio of all risky assets, including both private and public equity. Denote it by

$$
r_M = w_P^{\mathsf T}r_P + w_U^{\mathsf T}r_U,
$$

where $w_P$ and $w_U$ are the assets' market-value weights and their elements sum jointly to one. The standard CAPM consequently holds for every public or private asset $i$:

$$
E[r_i]-r_f
= \beta_{i,M}\bigl(E[r_M]-r_f\bigr),
\qquad
\beta_{i,M}=\frac{\operatorname{Cov}(r_i,r_M)}{\operatorname{Var}(r_M)}.
$$

Thus, a CAPM using only a public-stock index is generally not the correct model in this benchmark. Such an index omits private equity and is the relevant market portfolio only in the special case in which it is itself mean-variance equivalent to the aggregate portfolio of all risky assets.

Therefore, the standard CAPM holds, and its relevant market portfolio is the value-weighted portfolio of all private and public equity.

## (b) Investor (demand-side) segmentation

The PE fund can invest only in private equity and the risk-free asset, whereas the equity mutual fund can invest only in public equity and the risk-free asset. Hence, the two investors face different risky investment opportunity sets and choose different tangency portfolios.

The PE fund generates all demand for private equity. It combines the risk-free asset with the tangency portfolio formed from private assets. Because no other investor can hold private equity, market clearing requires this tangency portfolio to equal the value-weighted portfolio of all private equity, denoted by $r_{M_P}$.

Similarly, the mutual fund generates all demand for public equity. Its tangency portfolio is formed only from public assets, and market clearing requires it to equal the value-weighted portfolio of all public equity, denoted by $r_{M_U}$.

A separate CAPM therefore holds within each asset class. For every private asset $i\in P$,

$$
E[r_i]-r_f
=\beta_{i,M_P}\bigl(E[r_{M_P}]-r_f\bigr),
\qquad
\beta_{i,M_P}
=\frac{\operatorname{Cov}(r_i,r_{M_P})}
{\operatorname{Var}(r_{M_P})}.
$$

For every public asset $j\in U$,

$$
E[r_j]-r_f
=\beta_{j,M_U}\bigl(E[r_{M_U}]-r_f\bigr),
\qquad
\beta_{j,M_U}
=\frac{\operatorname{Cov}(r_j,r_{M_U})}
{\operatorname{Var}(r_{M_U})}.
$$

Thus, there is generally no single CAPM with one economy-wide market portfolio that prices both classes. The relevant benchmark for private equity is the private-equity market portfolio, while the relevant benchmark for public equity is the public-equity market portfolio. The cross-covariance block $\Sigma_{PU}$ does not affect either investor's portfolio choice because neither investor can hold assets from both classes. Consequently, the two classes can have different prices of risk, determined separately by demand and supply within each segmented market.

Therefore, two class-specific CAPMs hold: $M_P$ prices private equity and $M_U$ prices public equity, while no single economy-wide CAPM generally holds.

## (c) Uncorrelated asset classes and imperfect investor segmentation

The PE fund can now invest in both private and public equity, whereas the equity mutual fund remains restricted to public equity. Because $\Sigma_{PU}=0$, the covariance matrix and its inverse are block diagonal:

$$
\Sigma=
\begin{pmatrix}
\Sigma_P & 0\\
0 & \Sigma_U
\end{pmatrix},
\qquad
\Sigma^{-1}=
\begin{pmatrix}
\Sigma_P^{-1} & 0\\
0 & \Sigma_U^{-1}
\end{pmatrix}.
$$

Let $\mu_P=E[r_P]-r_f\mathbf{1}_P$ and $\mu_U=E[r_U]-r_f\mathbf{1}_U$ denote the vectors of expected excess returns. Mean-variance demand is proportional to $\Sigma^{-1}\mu$. If $\tau_{PE}$ and $\tau_{MF}$ denote the investors' risk tolerances, their risky-asset demands are therefore

$$
x_P^{PE}=\tau_{PE}\Sigma_P^{-1}\mu_P,
\qquad
x_U^{PE}=\tau_{PE}\Sigma_U^{-1}\mu_U,
$$

$$
x_P^{MF}=0,
\qquad
x_U^{MF}=\tau_{MF}\Sigma_U^{-1}\mu_U.
$$

The zero cross-covariance makes the PE fund's private- and public-equity demands separable: adding public equity does not hedge or alter the demand for private equity, and vice versa.

Let $s_P$ and $s_U$ denote the outstanding supplies of the individual assets in the two classes. Market clearing gives

$$
s_P=x_P^{PE}
=\tau_{PE}\Sigma_P^{-1}\mu_P,
$$

because only the PE fund holds private equity, and

$$
s_U=x_U^{PE}+x_U^{MF}
=(\tau_{PE}+\tau_{MF})\Sigma_U^{-1}\mu_U,
$$

because both investors hold public equity. Consequently,

$$
\mu_P=\frac{1}{\tau_{PE}}\Sigma_Ps_P,
\qquad
\mu_U=\frac{1}{\tau_{PE}+\tau_{MF}}\Sigma_Us_U.
$$

Within private equity, expected returns thus satisfy a CAPM relative to the value-weighted private-equity market portfolio $M_P$:

$$
E[r_i]-r_f
=\beta_{i,M_P}\bigl(E[r_{M_P}]-r_f\bigr),
\qquad i\in P.
$$

Within public equity, expected returns satisfy a CAPM relative to the value-weighted public-equity market portfolio $M_U$:

$$
E[r_j]-r_f
=\beta_{j,M_U}\bigl(E[r_{M_U}]-r_f\bigr),
\qquad j\in U.
$$

Both investors choose the same relative composition within their public-equity holdings, so their aggregate demand reproduces the public-equity market portfolio. The PE fund alone holds the entire private-equity supply, so its private-equity sleeve reproduces the private-equity market portfolio.

In general, there is still no single CAPM based on the value-weighted portfolio of all public and private equity. Private-equity risk is borne only by the PE fund, whereas public-equity risk is shared by both investors. This difference generally produces different equilibrium compensation for risk across the two classes. A global CAPM would arise only under an additional knife-edge condition that aligns the two class-specific risk prices.

Therefore, separate class-specific CAPMs hold, with $M_P$ as the benchmark for private assets and $M_U$ as the benchmark for public assets, but the aggregate market portfolio does not generally support a standard CAPM.

## (d) Imperfect investor segmentation with correlated asset classes

As in part (c), the PE fund can invest in both private and public equity, while the equity mutual fund can invest only in public equity. The difference is that $\Sigma_{PU}\neq 0$, so public equity can now hedge or amplify the risk of the PE fund's private-equity holdings.

Let $x_P^{PE}$ and $x_U^{PE}$ denote the PE fund's holdings and let $x_U^{MF}$ denote the mutual fund's public-equity holdings. Their mean-variance first-order conditions can be written as

$$
\mu_P
=a_{PE}\left(\Sigma_Px_P^{PE}+\Sigma_{PU}x_U^{PE}\right),
$$

$$
\mu_U
=a_{PE}\left(\Sigma_{PU}^{\mathsf T}x_P^{PE}+\Sigma_Ux_U^{PE}\right)
=a_{MF}\Sigma_Ux_U^{MF},
$$

where $\mu_P$ and $\mu_U$ are the vectors of expected excess returns and $a_{PE}$ and $a_{MF}$ are the investors' risk-aversion coefficients. Solving the public-equity conditions for the two demands gives

$$
x_U^{PE}
=\frac{1}{a_{PE}}\Sigma_U^{-1}\mu_U
-\Sigma_U^{-1}\Sigma_{PU}^{\mathsf T}x_P^{PE},
$$

$$
x_U^{MF}
=\frac{1}{a_{MF}}\Sigma_U^{-1}\mu_U.
$$

The mutual fund chooses public assets only for their public-equity risk-return trade-off. The PE fund also uses public assets to hedge its private-equity position, as shown by the additional term involving $\Sigma_{PU}^{\mathsf T}x_P^{PE}$. Therefore, the two investors generally do not hold the same relative weights within public equity.

Market clearing requires

$$
s_P=x_P^{PE},
\qquad
s_U=x_U^{PE}+x_U^{MF}.
$$

The public-equity market portfolio is consequently a mixture of two differently composed public portfolios. It need not coincide with the public-only tangency portfolio of the mutual fund or with the public-equity sleeve of the PE fund. Likewise, although the PE fund holds the entire private-equity supply, its optimal private holdings depend on their covariance with its public holdings. The private-equity market portfolio alone is therefore not generally a tangency portfolio either.

The usual CAPM equilibrium argument now breaks down. In the standard CAPM, every investor holds the same risky tangency portfolio, so adding their demands makes that tangency portfolio equal the value-weighted market portfolio. Here, the investors hold different risky portfolios. Their aggregate holdings reproduce asset supply, but aggregation does not produce a common tangency portfolio. Thus, neither the value-weighted public-equity portfolio nor the value-weighted portfolio of all risky assets is generally mean-variance efficient, and neither generally serves as a valid CAPM benchmark.

A portfolio-specific CAPM relation still holds for each investor's own opportunity set. If $T_{PE}$ is the PE fund's tangency portfolio formed from both asset classes, then

$$
E[r_i]-r_f
=\beta_{i,T_{PE}}\bigl(E[r_{T_{PE}}]-r_f\bigr),
\qquad i\in P\cup U.
$$

If $T_{MF}$ is the mutual fund's public-only tangency portfolio, then

$$
E[r_j]-r_f
=\beta_{j,T_{MF}}\bigl(E[r_{T_{MF}}]-r_f\bigr),
\qquad j\in U.
$$

These are investor-specific tangency-portfolio relations rather than a standard equilibrium CAPM with an observable market portfolio.

Therefore, the standard market CAPM generally does not hold: $T_{PE}$ prices the assets available to the PE fund, $T_{MF}$ prices the public assets available to the mutual fund, and there is no unique market portfolio that prices every asset.

## (e) Listed PE fund

The PE fund is now publicly listed, and its shares trade at net asset value. A share in the fund is therefore a transparent claim on the fund's underlying private- and public-equity holdings. If $r_L$ denotes the return on the listed fund, then

$$
r_L
=w_{P,L}^{\mathsf T}r_P
+w_{U,L}^{\mathsf T}r_U,
$$

where $w_{P,L}$ and $w_{U,L}$ are the weights of private and public assets held by the fund.

Because the PE fund can invest in both asset classes, its optimal risky portfolio is the tangency portfolio for the full universe of private and public equity. Listing the fund makes the return on this full-universe tangency portfolio directly available to the mutual fund. Direct public stocks cannot improve on its Sharpe ratio because they were already included in the larger opportunity set from which the PE fund's tangency portfolio was formed.

The mutual fund therefore also chooses the listed PE fund as its risky tangency portfolio and combines it with the risk-free asset. Both investors now hold the same composition of risky wealth. The common tangency-portfolio argument is restored, and market clearing identifies this tangency portfolio with the value-weighted portfolio of all risky assets.

The standard CAPM therefore holds:

$$
E[r_i]-r_f
=\beta_{i,M}\bigl(E[r_M]-r_f\bigr),
\qquad
\beta_{i,M}
=\frac{\operatorname{Cov}(r_i,r_M)}
{\operatorname{Var}(r_M)}.
$$

The relevant market portfolio can be described in two equivalent ways:

1. On a look-through basis, it is the value-weighted portfolio of all underlying private and public equity, with every underlying asset counted once.
2. In terms of securities available to the mutual fund, it is the value-weighted portfolio of the listed PE fund and the public equity held outside that fund.

The second description must not add the PE fund's full market capitalization to the full market capitalization of all public companies without an adjustment. Doing so would count the public shares owned inside the PE fund twice. At net asset value, the correctly adjusted securities-level portfolio and the look-through portfolio have exactly the same payoff.

Hence, unlike in part (d), a public-stock benchmark can represent the complete market portfolio if it includes the listed PE fund and is adjusted for its underlying cross-holdings. The assumption that the fund trades at net asset value is important: a discount, premium, management-cost wedge, or other fund-specific friction would break the exact equivalence between the listed claim and its underlying portfolio.

Therefore, the standard CAPM holds, and the relevant market portfolio is all private and public equity counted once, equivalently represented by the listed PE fund together with public equity held outside the fund.

## (f) Implications for applying the CAPM

Segmentation changes which risks investors can trade and which investors are marginal in each market. Consequently, the appropriate CAPM benchmark need not be the value-weighted portfolio of every risky asset in the economy. It should instead be a mean-variance-efficient portfolio for the opportunity set relevant to the investors who price the assets under consideration.

It is therefore not always a bad idea to use a public-stock index when pricing public equity. Its validity depends on the economic setting:

- In the frictionless benchmark of part (a), investors can hold both asset classes. The relevant market portfolio contains all public and private equity, so a public-only index generally omits priced risk and is not the correct CAPM benchmark.
- Under complete segmentation in part (b), public equity is priced within its own market. The value-weighted public-equity portfolio is the relevant benchmark for public stocks even though it is not the economy-wide wealth portfolio.
- In part (c), the asset classes are uncorrelated. Both investors choose the same relative public-equity weights, so their aggregate public holdings equal the public tangency portfolio. A public-stock index is again an appropriate benchmark for public equity.
- In part (d), private and public returns are correlated. The PE fund uses public assets partly to hedge private-equity risk, while the mutual fund does not. Their public holdings therefore differ, and the aggregate public-stock index need not be a tangency portfolio. In this case, using it as a CAPM benchmark can be misleading.
- In part (e), listing the PE fund at net asset value makes private-equity exposure indirectly tradable. The standard CAPM is restored, but the public benchmark must include the listed PE fund and correctly account for its underlying holdings.

The main lesson is that the quality of a market proxy is determined by investor access, spanning, and the identity of the marginal risk bearers, not simply by whether the proxy is publicly observable. A public-stock index can price public equity well when it represents the relevant investors' tangency portfolio or spans the omitted risks. It can fail when excluded assets affect those investors' hedging demands or when the index leaves out a material component of their investable aggregate wealth.

Therefore, using a public-stock index is not always a bad idea: it is correct when that index is the relevant public-market tangency portfolio or spans the complete market, and it is incorrect when segmentation and cross-asset hedging prevent it from being mean-variance efficient.

## Question 1: Changing portfolio weights (qualitative)

### Background
**(a) Motivation for changing weights.** SF runs a **tactical asset allocation (market timing)** strategy around a 60/40 stock/T-bill benchmark, within a 40% to 80% equity range. It overweighted equities in 2006 and 2012 to 2013 (65% today), cut towards 40% in 2007 to 2008, and has beaten the benchmark by 0.98% per year since 2006 (Exhibit 3). With mean-variance preferences the optimal risky share is $y^* = \frac{E[r_M]-r_f}{A\,\sigma_M^2}$, so changing views on the equity premium or on market risk change $y^*$.

**(b) Reward to risk along the CAL.** With a risk-free asset and one risky portfolio, a mix with weight $y$ in stocks has
$E[r_p]-r_f = y\,(E[r_M]-r_f)$ and $\sigma_p = y\,\sigma_M$, so
$$S_p = \frac{E[r_p]-r_f}{\sigma_p} = \frac{E[r_M]-r_f}{\sigma_M} = S_M ,$$
independent of $y$.

### Answers: Question 1

**(a)**

The main motivation for changing the portfolio weights is market timing. Schumpeter's manager shifts the allocation between the stock index and T-bills based on her view of where the market is heading, overweighting equities when she expects the market to rise (as in 2006 and 2012 to 2013, and with 65% today) and moving towards T-bills when she expects a fall (cutting the equity share towards 40% during 2007 and 2008). If these views are right, the fund earns the equity premium in good periods while avoiding part of the losses in bad ones. This is what happened in 2008, when the fund lost 26.8% compared to 33.3% for the benchmark and 46% for the equity market.

A second motivation is differentiation. A fixed 60/40 portfolio can be bought from any investment firm, so SF has to offer something more to attract investors. The strategy does not need to outperform the whole market in order to be seen as successful, simply outperforming a conservative 60/40 index/T-bill portfolio is enough, and since 2006 SF has done so by 0.98% per year.

Finally, the strategy fits SF's size. Stock picking requires research resources that a small firm with only one investment professional does not have, while market timing only requires trading a liquid index and government bills. This makes the strategy easy to implement, with low trading costs when rebalancing (around 7 bp per year), and possible to run with a small amount of AUM.

**(b)**

Under mean-variance theory the other manager is right. With a risk-free asset and a single risky portfolio, every mix of the stock index and T-bills lies on the Capital Allocation Line. A weight $y$ in the index gives an expected excess return of $E[r_p]-r_f = y\,(E[r_M]-r_f)$ and a standard deviation of $\sigma_p = y\,\sigma_M$, so the Sharpe ratio of the mix is

$$S_p = \frac{y\,(E[r_M]-r_f)}{y\,\sigma_M} = \frac{E[r_M]-r_f}{\sigma_M} = S_M ,$$

which does not depend on $y$. Changing the weight only moves the portfolio along the CAL, taking on more or less risk at the same reward per unit of risk, and cannot make the line any steeper.

This only holds if there is no alpha and expected returns are constant and already priced in. If not all information is priced in, a manager who increases the equity weight before a period of higher expected returns (and reduces it before a period of lower ones) is in effect moving between CALs whose slope changes over time. The realised reward to risk of the timing strategy can then exceed that of any fixed mix, such as the 60/40 benchmark. Schumpeter's 0.98% yearly outperformance since 2006 is consistent with this, but it only adds value if the timing is based on real information rather than luck.

### 2b and 2c. Portfolio standard deviations
Base case: 65% index, 35% T-bills. Then three ways of moving 1% out of T-bills.

The portfolio SD is computed directly from the time series of portfolio excess returns.

```python
portfolios = {
    "Base: 65% STOXX / 35% T-bills":           {"STOXX": 0.65, "TKA": 0.00, "OLE": 0.00},
    "(i)   66% STOXX / 34% T-bills":           {"STOXX": 0.66, "TKA": 0.00, "OLE": 0.00},
    "(ii)  65% STOXX + 1% Deoleo / 34% T-bills":      {"STOXX": 0.65, "TKA": 0.00, "OLE": 0.01},
    "(iii) 65% STOXX + 1% ThyssenKrupp / 34% T-bills": {"STOXX": 0.65, "TKA": 0.01, "OLE": 0.00},
}
W = pd.DataFrame(portfolios).T[assets]
W.insert(0, "T-bills", 1 - W.sum(axis=1))

port_ex = excess[assets] @ W[assets].T          # T x 4 matrix of portfolio excess returns
port_sd = port_ex.std() * np.sqrt(12)

base = port_sd.iloc[0]
q2 = W.copy()
q2["Annualized SD"] = port_sd
q2["Change vs base (pp)"] = (port_sd - base) * 100
q2["Change vs base (bp)"] = (port_sd - base) * 1e4

Sigma_a = excess[assets].cov() * 12            # annualized covariance matrix (used in 2d)

q2.style.format({"T-bills": "{:.0%}", "STOXX": "{:.0%}", "TKA": "{:.0%}", "OLE": "{:.0%}",
                 "Annualized SD": "{:.3%}", "Change vs base (pp)": "{:+.3f}",
                 "Change vs base (bp)": "{:+.1f}"})
```

### 2d. Why does the stand-alone risk ranking from 2a not predict the ranking in 2c? A risk decomposition

The variance of the portfolio is $\sigma_p^2 = w^\top\Sigma w = \sum_i w_i\,\text{Cov}(r_i, r_p)$.
So asset $i$'s **contribution** to portfolio variance is $w_i\,\text{Cov}(r_i,r_p)$, and the **marginal effect** of adding a little of asset $i$ (funded from T-bills) is

$$\frac{\partial \sigma_p}{\partial w_i} = \frac{\text{Cov}(r_i, r_p)}{\sigma_p} = \rho_{i,p}\,\sigma_i .$$

What matters at the margin is therefore not $\sigma_i$ but only the part of it that is correlated with the portfolio, $\rho_{i,p}\sigma_i$. Since the base portfolio is 65% of the index, $\text{Cov}(r_i,r_p) = 0.65\,\text{Cov}(r_i,r_M) = 0.65\,\beta_i\sigma_M^2$, so

$$\Delta\sigma_p \approx 0.01 \times \rho_{i,M}\,\sigma_i = 0.01 \times \beta_i\,\sigma_M .$$

Below: the correlations, the marginal risk of each asset, the linear prediction of $\Delta\sigma_p$, and the actual change from 2c.

```python
w0 = W.loc[W.index[0], assets]                 # base portfolio (65% STOXX)
sig_p = np.sqrt(w0 @ Sigma_a @ w0)
cov_ip = Sigma_a @ w0                           # Cov(r_i, r_p), annualized
corr_iM = excess[assets].corr()["STOXX"]

decomp = pd.DataFrame({
    "Stand-alone SD": sd_a,
    "Corr with index": corr_iM,
    "Correlated part (rho x SD)": corr_iM * sd_a,
    "Uncorrelated part (SD x sqrt(1-rho^2))": sd_a * np.sqrt(1 - corr_iM**2),
    "Predicted change, +1% (bp)": 0.01 * cov_ip / sig_p * 1e4,
})
# map each asset to its 2c portfolio: (i) index, (ii) Deoleo, (iii) ThyssenKrupp
chg_2c = pd.Series(q2["Change vs base (bp)"].iloc[1:].values, index=["STOXX", "OLE", "TKA"])
decomp["Actual change, +1% (bp)"] = chg_2c
decomp.rename(index=NAMES).T
```

Visual version: each asset's stand-alone SD split into the part correlated with the market (which adds to portfolio risk at the margin) and the uncorrelated part (which is diversified away for a small position).

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4.2))
lab = [NAMES[a] for a in assets]
x = np.arange(len(assets))

ax = axes[0]
corr_part = (corr_iM * sd_a) * 100
unc_part = (sd_a - corr_iM * sd_a) * 100
ax.bar(x, corr_part, color=[COL[a] for a in assets], width=0.6, label="Correlated with index ($\\rho\\sigma_i$)")
ax.bar(x, unc_part, bottom=corr_part, color=[COL[a] for a in assets], alpha=0.3, width=0.6,
       label="Rest of stand-alone SD")
for i, a in enumerate(assets):
    ax.text(i, sd_a[a] * 100 + 1, f"{sd_a[a]:.1%}", ha="center", fontsize=9)
    ax.text(i, corr_part[a] / 2, f"{corr_part[a]:.1f}%", ha="center", fontsize=9, color="white")
ax.set_xticks(x, lab); ax.set_ylabel("Annualized SD (%)")
from matplotlib.patches import Patch
ax.set_title("2a: stand-alone risk", loc="left")
ax.legend(handles=[Patch(color="dimgray", label="Correlated with index ($\\rho_{i,M}\\sigma_i$)"),
                   Patch(color="dimgray", alpha=0.3, label="Rest of stand-alone SD (diversifiable)")],
          frameon=False, fontsize=8, loc="upper left")

ax = axes[1]
chg = chg_2c[assets].values
ax.bar(x, chg, color=[COL[a] for a in assets], width=0.6)
for i, v in enumerate(chg):
    ax.text(i, v + 0.3, f"{v:+.1f} bp", ha="center", fontsize=9)
ax.set_xticks(x, lab); ax.set_ylabel("Change in portfolio SD (bp)")
ax.set_title("2c: effect of moving 1% from T-bills", loc="left")
plt.tight_layout(); plt.show()
```

### Answers: Question 2

**(d)**

Although the stand-alone standard deviations from 2a suggest that the STOXX index is the least risky asset and Deoleo the most risky, the results in 2c show that adding 1% of Deoleo increases the portfolio standard deviation the least: 13.0 bp, compared to 15.7 bp for the index and 27.2 bp for ThyssenKrupp. The reason is that the risk an asset adds to a portfolio does not depend on its stand-alone risk, but on how it moves together with the rest of the portfolio, which here is 65% the STOXX index.

Deoleo has a correlation with the index of only 0.25, so only about 11.9% of its 48.4% volatility moves with the portfolio. The rest is idiosyncratic and is diversified away when the position is small. ThyssenKrupp on the other hand has a correlation of 0.67, meaning that 26.8% of its 39.7% volatility moves with the index, which makes it the most costly addition in terms of risk.

A simple risk decomposition makes this concrete. The marginal effect on portfolio standard deviation of adding a small amount of asset $i$ is $\text{Cov}(r_i,r_p)/\sigma_p = \rho_{i,p}\,\sigma_i$, so moving 1% into asset $i$ should increase the portfolio standard deviation by roughly $0.01 \times \rho_{i,M}\,\sigma_i$. This gives 15.7, 26.8 and 11.9 bp, which is very close to the actual changes. The small gap comes from a second-order term ($w_i^2\sigma_i^2$) that grows with the size of the position, so the diversification benefit of Deoleo would shrink if the fund bought a larger stake.

### 3b. Link to Question 2
A stock's variance splits into a systematic part $\beta_i^2\sigma_M^2$ and an idiosyncratic part $\sigma^2_{\varepsilon_i}$. For a small position added to a market-like portfolio, the increase in portfolio SD is approximately $\Delta w\,\beta_i\,\sigma_M$ (since $\beta_i\sigma_M = \rho_{i,M}\sigma_i$, this is the same prediction as in 2d). Idiosyncratic risk only enters at second order ($\Delta w^2\sigma^2_{\varepsilon_i}$).
